## 合并全局变量_2017

```python
import pandas as pd
import os

# 1. 先读入 transmod（如果你已经有了 DataFrame 就跳过这一步）
transmod = pd.read_csv(r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\2017\transmod_2017.csv')

# 2. 所有变量文件所在的目录
base_path = r"C:\Users\Fe6ruar9\PhD\313.5-Heatwave\2017\variables"

# 3. 变量名列表（须与 CSV 里对应列完全一致）
variables = ['AirFV', 'AirPV', 'GDP', 'NDVI', 'population',
             'RailFV', 'RailPV', 'RoadFV', 'RoadPV']

# 4. 循环合并
for var in variables:
    file_path = os.path.join(base_path, f"{var}_2017.csv")
    df_var = pd.read_csv(file_path)                 # 读取：包含 city_name 和 var 列
    
    # origin 合并，后缀 _o
    transmod = (
        transmod
        .merge(
            df_var[['city_name', var]],
            how='left',
            left_on='origin',
            right_on='city_name'
        )
        .rename(columns={var: f"{var}_o"})
        .drop(columns=['city_name'])
    )
    
    # dest 合并，后缀 _d
    transmod = (
        transmod
        .merge(
            df_var[['city_name', var]],
            how='left',
            left_on='dest',
            right_on='city_name'
        )
        .rename(columns={var: f"{var}_d"})
        .drop(columns=['city_name'])
    )

# 5. （可选）检查结果
print(transmod.head())

# 6. （可选）保存合并后的 DataFrame
transmod.to_csv(r"C:\Users\Fe6ruar9\PhD\313.5-Heatwave\2017\transmod_enriched.csv", index=False)
```

### 通过gdp population来填补缺失值

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 假设您的数据框已经加载为 data_intensity_2024
# data_intensity_2024 = pd.read_csv('your_file.csv')

def analyze_and_impute_traffic_data(df, method='random_forest'):
    """
    分析交通变量与GDP、人口的关系并填补缺失值
    
    参数:
    df: pandas DataFrame
    method: 'linear' 或 'random_forest'
    """
    
    # 定义需要填补的变量
    traffic_vars = [
        'AirFV_o', 'AirFV_d', 'AirPV_o', 'AirPV_d',
        'RailFV_o', 'RailFV_d', 'RailPV_o', 'RailPV_d',
        'RoadFV_o', 'RoadFV_d', 'RoadPV_o', 'RoadPV_d'
    ]
    
    # 创建数据副本
    df_imputed = df.copy()
    
    # 存储模型性能指标
    model_performance = {}
    
    # 对每个交通变量进行分析和填补
    for var in traffic_vars:
        print(f"\n处理变量: {var}")
        
        # 确定是origin还是destination
        suffix = var.split('_')[-1]
        gdp_var = f'GDP_{suffix}'
        pop_var = f'population_{suffix}'
        
        # 准备训练数据（非缺失值）
        mask_not_null = df[var].notna()
        
        if mask_not_null.sum() == 0:
            print(f"警告: {var} 全部为缺失值，跳过")
            continue
            
        # 特征变量
        feature_cols = [gdp_var, pop_var]
        
        # 可以添加更多相关特征
        if 'NDVI_' + suffix in df.columns:
            feature_cols.append('NDVI_' + suffix)
        
        # 训练数据
        X_train = df.loc[mask_not_null, feature_cols]
        y_train = df.loc[mask_not_null, var]
        
        # 检查是否有足够的训练数据
        if len(X_train) < 10:
            print(f"警告: {var} 的非缺失值太少 ({len(X_train)}个)，跳过")
            continue
        
        # 分割训练集和验证集
        if len(X_train) > 50:
            X_tr, X_val, y_tr, y_val = train_test_split(
                X_train, y_train, test_size=0.2, random_state=42
            )
        else:
            X_tr, y_tr = X_train, y_train
            X_val, y_val = None, None
        
        # 选择模型
        if method == 'linear':
            model = LinearRegression()
        else:  # random_forest
            model = RandomForestRegressor(
                n_estimators=100,
                max_depth=10,
                random_state=42
            )
        
        # 训练模型
        model.fit(X_tr, y_tr)
        
        # 评估模型
        if X_val is not None:
            y_pred = model.predict(X_val)
            mse = mean_squared_error(y_val, y_pred)
            r2 = r2_score(y_val, y_pred)
            print(f"  验证集 R²: {r2:.3f}, RMSE: {np.sqrt(mse):.3f}")
            model_performance[var] = {'r2': r2, 'rmse': np.sqrt(mse)}
        
        # 预测缺失值
        mask_null = df[var].isna()
        if mask_null.sum() > 0:
            X_predict = df.loc[mask_null, feature_cols]
            
            # 处理预测数据中的缺失值
            if X_predict.isna().any().any():
                print(f"  警告: 预测特征中存在缺失值，使用均值填充")
                X_predict = X_predict.fillna(X_predict.mean())
            
            # 进行预测
            predictions = model.predict(X_predict)
            
            # 确保预测值非负
            predictions = np.maximum(predictions, 0)
            
            # 填补缺失值
            df_imputed.loc[mask_null, var] = predictions
            print(f"  已填补 {mask_null.sum()} 个缺失值")
    
    return df_imputed, model_performance

def visualize_relationships(df):
    """
    可视化交通变量与GDP、人口的关系
    """
    # 选择几个代表性变量进行可视化
    sample_vars = ['AirPV_o', 'RailPV_o', 'RoadFV_o']
    
    fig, axes = plt.subplots(3, 2, figsize=(12, 15))
    axes = axes.flatten()
    
    plot_idx = 0
    for var in sample_vars:
        if var not in df.columns:
            continue
            
        suffix = var.split('_')[-1]
        gdp_var = f'GDP_{suffix}'
        pop_var = f'population_{suffix}'
        
        # GDP关系图
        ax = axes[plot_idx]
        mask = df[var].notna() & df[gdp_var].notna()
        ax.scatter(df.loc[mask, gdp_var], df.loc[mask, var], alpha=0.5)
        ax.set_xlabel(gdp_var)
        ax.set_ylabel(var)
        ax.set_title(f'{var} vs {gdp_var}')
        plot_idx += 1
        
        # 人口关系图
        ax = axes[plot_idx]
        mask = df[var].notna() & df[pop_var].notna()
        ax.scatter(df.loc[mask, pop_var], df.loc[mask, var], alpha=0.5)
        ax.set_xlabel(pop_var)
        ax.set_ylabel(var)
        ax.set_title(f'{var} vs {pop_var}')
        plot_idx += 1
    
    plt.tight_layout()
    plt.show()

def analyze_missing_patterns(df):
    """
    分析缺失值模式
    """
    traffic_vars = [
        'AirFV_o', 'AirFV_d', 'AirPV_o', 'AirPV_d',
        'RailFV_o', 'RailFV_d', 'RailPV_o', 'RailPV_d',
        'RoadFV_o', 'RoadFV_d', 'RoadPV_o', 'RoadPV_d'
    ]
    
    # 计算缺失值统计
    missing_stats = pd.DataFrame({
        'Variable': traffic_vars,
        'Missing_Count': [df[var].isna().sum() for var in traffic_vars],
        'Missing_Percentage': [df[var].isna().sum() / len(df) * 100 for var in traffic_vars],
        'Non_Missing_Count': [df[var].notna().sum() for var in traffic_vars]
    })
    
    print("缺失值统计:")
    print(missing_stats)
    
    # 可视化缺失值模式
    plt.figure(figsize=(10, 6))
    missing_matrix = df[traffic_vars].isna().astype(int)
    sns.heatmap(missing_matrix.T, cmap='RdYlBu', cbar_kws={'label': '缺失值'})
    plt.xlabel('样本索引')
    plt.ylabel('变量')
    plt.title('缺失值模式热图')
    plt.tight_layout()
    plt.show()
    
    return missing_stats

# 主程序
def main(data_intensity_2024):
    """
    主函数：执行完整的缺失值分析和填补流程
    """
    print("=== 开始分析交通数据缺失值 ===")
    
    # 1. 分析缺失值模式
    print("\n1. 分析缺失值模式")
    missing_stats = analyze_missing_patterns(data_intensity_2024)
    
    # 2. 可视化变量关系
    print("\n2. 可视化交通变量与GDP、人口的关系")
    visualize_relationships(data_intensity_2024)
    
    # 3. 使用随机森林填补缺失值
    print("\n3. 使用随机森林模型填补缺失值")
    df_rf, performance_rf = analyze_and_impute_traffic_data(
        data_intensity_2024, method='random_forest'
    )
    
    # 4. 也可以尝试线性回归（作为对比）
    print("\n4. 使用线性回归模型填补缺失值（对比）")
    df_linear, performance_linear = analyze_and_impute_traffic_data(
        data_intensity_2024, method='linear'
    )
    
    # 5. 比较模型性能
    print("\n5. 模型性能比较")
    performance_comparison = pd.DataFrame({
        'Random_Forest_R2': [performance_rf.get(var, {}).get('r2', np.nan) 
                             for var in performance_rf.keys()],
        'Linear_R2': [performance_linear.get(var, {}).get('r2', np.nan) 
                      for var in performance_linear.keys()]
    }, index=performance_rf.keys())
    print(performance_comparison)
    
    # 6. 验证填补结果
    print("\n6. 填补后的缺失值统计")
    traffic_vars = [
        'AirFV_o', 'AirFV_d', 'AirPV_o', 'AirPV_d',
        'RailFV_o', 'RailFV_d', 'RailPV_o', 'RailPV_d',
        'RoadFV_o', 'RoadFV_d', 'RoadPV_o', 'RoadPV_d'
    ]
    
    for var in traffic_vars:
        original_missing = data_intensity_2024[var].isna().sum()
        rf_missing = df_rf[var].isna().sum()
        print(f"{var}: 原始缺失 {original_missing} -> 填补后缺失 {rf_missing}")
    
    return df_rf, df_linear, performance_comparison

# 使用示例
# 假设您已经加载了数据
# df_imputed_rf, df_imputed_linear, performance = main(data_intensity_2024)

# 如果只想快速填补缺失值
# df_imputed, _ = analyze_and_impute_traffic_data(data_intensity_2024, method='random_forest')

# 保存填补后的数据
# df_imputed.to_csv('data_intensity_2024_imputed.csv', index=False)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, PolynomialFeatures, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, log_loss, make_scorer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
import xgboost as xgb
import warnings
warnings.filterwarnings('ignore')


class EnhancedTransportModeModel:
    """增强版交通模式选择模型"""
    
    def __init__(self, model_type='xgboost', random_state=42):
        """
        Parameters:
        -----------
        model_type : str
            模型类型: 'logistic', 'random_forest', 'xgboost', 'neural_network', 'ensemble'
        """
        self.model_type = model_type
        self.model = None
        self.scaler = StandardScaler()
        self.imputer = None
        self.random_state = random_state
        self.feature_names = None
        self.enhanced_features = None
        self.classes_ = ['car', 'train', 'plane']
        self.label_encoder = LabelEncoder()
        self.label_encoder.fit(self.classes_)  # 预设标签编码器
        
    def create_features(self, df):
        """创建增强特征"""
        data = df.copy()
        
        # 1. 距离相关特征（如果有origin和dest）
        if 'origin' in data.columns and 'dest' in data.columns:
            # 这里可以加入实际距离计算
            pass
        
        # 2. 基础设施差异特征
        data['AirFV_diff'] = data['AirFV_o'] - data['AirFV_d']
        data['RailFV_diff'] = data['RailFV_o'] - data['RailFV_d']
        data['RoadFV_diff'] = data['RoadFV_o'] - data['RoadFV_d']
        
        # 3. 基础设施比率特征
        data['Air_vs_Rail_o'] = data['AirFV_o'] / (data['RailFV_o'] + 1e-6)
        data['Air_vs_Road_o'] = data['AirFV_o'] / (data['RoadFV_o'] + 1e-6)
        data['Rail_vs_Road_o'] = data['RailFV_o'] / (data['RoadFV_o'] + 1e-6)
        
        # 4. 经济发展差异
        data['GDP_diff'] = data['GDP_o'] - data['GDP_d']
        data['GDP_ratio'] = data['GDP_o'] / (data['GDP_d'] + 1e-6)
        data['GDP_mean'] = (data['GDP_o'] + data['GDP_d']) / 2
        
        # 5. 人口特征
        data['population_diff'] = data['population_o'] - data['population_d']
        data['population_ratio'] = data['population_o'] / (data['population_d'] + 1e-6)
        data['population_density_o'] = data['population_o'] / (data['NDVI_o'] + 1e-6)
        data['population_density_d'] = data['population_d'] / (data['NDVI_d'] + 1e-6)
        
        # 6. 天气条件综合指标（移除SNDP）
        data['weather_severity_o'] = (
            data['PREC_o'] * 0.4 +  # 提高降水权重
            np.abs(data['temax_o'] - 20) * 0.3 +  # 偏离舒适温度
            data['WDSP_o'] * 0.3  # 风速权重
        )
        data['weather_severity_d'] = (
            data['PREC_d'] * 0.4 + 
            np.abs(data['temax_d'] - 20) * 0.3 +
            data['WDSP_d'] * 0.3
        )
        
        # 7. 交通基础设施综合指数
        data['transport_index_o'] = (
            data['AirPV_o'] * 0.4 + 
            data['RailPV_o'] * 0.3 + 
            data['RoadPV_o'] * 0.3
        )
        data['transport_index_d'] = (
            data['AirPV_d'] * 0.4 + 
            data['RailPV_d'] * 0.3 + 
            data['RoadPV_d'] * 0.3
        )
        
        # 8. 时间特征组合
        data['workday_weather_o'] = data['is_workday?'] * data['weather_severity_o']
        data['holiday_GDP'] = data['is_holiday?'] * data['GDP_mean']
        
        # 9. 非线性变换
        data['log_GDP_o'] = np.log1p(data['GDP_o'])
        data['log_GDP_d'] = np.log1p(data['GDP_d'])
        data['log_population_o'] = np.log1p(data['population_o'])
        data['log_population_d'] = np.log1p(data['population_d'])
        
        # 10. 交互特征
        data['GDP_population_o'] = data['GDP_o'] * data['population_o']
        data['GDP_population_d'] = data['GDP_d'] * data['population_d']
        
        # 保存新特征名
        new_features = [
            'AirFV_diff', 'RailFV_diff', 'RoadFV_diff',
            'Air_vs_Rail_o', 'Air_vs_Road_o', 'Rail_vs_Road_o',
            'GDP_diff', 'GDP_ratio', 'GDP_mean',
            'population_diff', 'population_ratio', 
            'population_density_o', 'population_density_d',
            'weather_severity_o', 'weather_severity_d',
            'transport_index_o', 'transport_index_d',
            'workday_weather_o', 'holiday_GDP',
            'log_GDP_o', 'log_GDP_d', 'log_population_o', 'log_population_d',
            'GDP_population_o', 'GDP_population_d'
        ]
        
        return data, new_features
    
    def prepare_data(self, transmod, test_size=0.2, use_enhanced_features=True):
        """准备数据"""
        # 基础特征（移除SNDP_o和SNDP_d）
        self.feature_names = [
            'AirFV_o', 'AirFV_d', 'AirPV_o', 'AirPV_d', 'GDP_o', 'GDP_d', 
            'NDVI_o', 'NDVI_d', 'population_o', 'population_d', 
            'RailFV_o', 'RailFV_d', 'RailPV_o', 'RailPV_d', 
            'RoadFV_o', 'RoadFV_d', 'RoadPV_o', 'RoadPV_d', 
            'PREC_o', 'PREC_d', 
            'temax_o', 'temax_d', 'WDSP_o', 'WDSP_d', 
            'is_workday?', 'is_weekend?', 'is_holiday?'
        ]
        
        # 创建增强特征
        if use_enhanced_features:
            data, new_features = self.create_features(transmod)
            self.enhanced_features = self.feature_names + new_features
            features_to_use = self.enhanced_features
        else:
            data = transmod.copy()
            features_to_use = self.feature_names
        
        # 处理缺失值
        self.imputer = SimpleImputer(strategy='median')
        data[features_to_use] = self.imputer.fit_transform(data[features_to_use])
        
        # 提取特征和目标
        X = data[features_to_use].values
        y_probs = data[['car', 'train', 'plane']].values
        y_probs = y_probs / y_probs.sum(axis=1, keepdims=True)
        
        # 修改：使用数值标签而不是字符串标签
        y_labels_numeric = y_probs.argmax(axis=1)  # 0, 1, 2
        y_labels_string = np.array([self.classes_[i] for i in y_labels_numeric])  # 用于显示
        
        # 划分数据集
        X_train, X_test, y_train_num, y_test_num, y_probs_train, y_probs_test = train_test_split(
            X, y_labels_numeric, y_probs, test_size=test_size, random_state=self.random_state, 
            stratify=y_labels_numeric
        )
        
        # 同时保存字符串标签用于评估显示
        y_train_str = np.array([self.classes_[i] for i in y_train_num])
        y_test_str = np.array([self.classes_[i] for i in y_test_num])
        
        # 标准化
        X_train_scaled = self.scaler.fit_transform(X_train)
        X_test_scaled = self.scaler.transform(X_test)
        
        # 返回数值标签用于训练，字符串标签用于显示
        return (X_train_scaled, X_test_scaled, y_train_num, y_test_num, 
                y_probs_train, y_probs_test, y_train_str, y_test_str)
    
    def create_model(self):
        """创建模型"""
        if self.model_type == 'logistic':
            # 增强的逻辑回归
            self.model = LogisticRegression(
                multi_class='multinomial',
                solver='saga',  # 支持L1正则化
                penalty='elasticnet',  # 弹性网络正则化
                l1_ratio=0.5,
                C=0.1,  # 正则化强度
                max_iter=2000,
                random_state=self.random_state
            )
            
        elif self.model_type == 'random_forest':
            self.model = RandomForestClassifier(
                n_estimators=300,
                max_depth=20,
                min_samples_split=50,
                min_samples_leaf=20,
                max_features='sqrt',
                random_state=self.random_state,
                n_jobs=-1
            )
            
        elif self.model_type == 'xgboost':
            self.model = xgb.XGBClassifier(
                n_estimators=500,
                max_depth=8,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                objective='multi:softprob',
                eval_metric='mlogloss',
                random_state=self.random_state,
                n_jobs=-1
            )
            
        elif self.model_type == 'neural_network':
            self.model = MLPClassifier(
                hidden_layer_sizes=(128, 64, 32),
                activation='relu',
                solver='adam',
                learning_rate='adaptive',
                learning_rate_init=0.001,
                max_iter=500,
                early_stopping=True,
                validation_fraction=0.1,
                random_state=self.random_state
            )
            
        elif self.model_type == 'ensemble':
            # 使用投票集成
            from sklearn.ensemble import VotingClassifier
            
            models = [
                ('lr', LogisticRegression(
                    multi_class='multinomial', solver='saga', 
                    penalty='elasticnet', l1_ratio=0.5, C=0.1,
                    max_iter=2000, random_state=self.random_state
                )),
                ('rf', RandomForestClassifier(
                    n_estimators=200, max_depth=15,
                    random_state=self.random_state, n_jobs=-1
                )),
                ('xgb', xgb.XGBClassifier(
                    n_estimators=300, max_depth=6, learning_rate=0.1,
                    random_state=self.random_state, n_jobs=-1
                ))
            ]
            
            self.model = VotingClassifier(
                estimators=models,
                voting='soft',  # 使用概率平均
                n_jobs=-1
            )
    
    def train_with_optimization(self, X_train, y_train, X_val=None, y_val=None):
        """带超参数优化的训练"""
        self.create_model()
        
        if self.model_type == 'xgboost' and X_val is not None:
            # XGBoost早停
            self.model.fit(
                X_train, y_train,
                eval_set=[(X_val, y_val)],
                early_stopping_rounds=50,
                verbose=False
            )
        else:
            self.model.fit(X_train, y_train)
            
    def evaluate_comprehensive(self, X_test, y_test_num, y_test_str, y_probs_test):
        """综合评估"""
        from scipy.stats import entropy
        
        # 预测
        y_pred_num = self.model.predict(X_test)
        y_pred_probs = self.model.predict_proba(X_test)
        
        # 转换为字符串标签用于报告
        y_pred_str = np.array([self.classes_[i] for i in y_pred_num])
        
        # 计算各种指标
        accuracy = accuracy_score(y_test_num, y_pred_num)
        logloss = log_loss(y_test_num, y_pred_probs)
        
        # 每个类别的准确率 - 使用字符串标签
        from sklearn.metrics import classification_report
        report = classification_report(y_test_str, y_pred_str, target_names=self.classes_, output_dict=True)
        
        # 概率预测的MAE
        mae = np.mean(np.abs(y_probs_test - y_pred_probs))
        rmse = np.sqrt(np.mean((y_probs_test - y_pred_probs) ** 2))
        
        # KL散度
        kl_divs = []
        for i in range(len(y_probs_test)):
            p = y_probs_test[i] + 1e-10
            q = y_pred_probs[i] + 1e-10
            kl_divs.append(entropy(p, q))
        avg_kl = np.mean(kl_divs)
        
        results = {
            'accuracy': accuracy,
            'log_loss': logloss,
            'mae': mae,
            'rmse': rmse,
            'kl_divergence': avg_kl,
            'per_class_precision': {k: report[k]['precision'] for k in self.classes_ if k in report},
            'per_class_recall': {k: report[k]['recall'] for k in self.classes_ if k in report}
        }
        
        return results, y_pred_probs


def compare_models(transmod):
    """比较不同模型的性能"""
    from scipy.stats import entropy
    
    # 准备数据（使用增强特征）
    base_model = EnhancedTransportModeModel()
    (X_train, X_test, y_train_num, y_test_num, 
     y_probs_train, y_probs_test, y_train_str, y_test_str) = base_model.prepare_data(
        transmod, use_enhanced_features=True
    )
    
    # 测试不同模型
    model_types = ['logistic', 'random_forest', 'xgboost', 'neural_network']
    results_summary = []
    
    for model_type in model_types:
        print(f"\n训练 {model_type} 模型...")
        model = EnhancedTransportModeModel(model_type=model_type)
        model.enhanced_features = base_model.enhanced_features
        model.scaler = base_model.scaler
        model.imputer = base_model.imputer
        
        # 训练 - 使用数值标签
        model.train_with_optimization(X_train, y_train_num)
        
        # 评估 - 传递数值和字符串标签
        results, _ = model.evaluate_comprehensive(X_test, y_test_num, y_test_str, y_probs_test)
        
        results_summary.append({
            '模型': model_type,
            '准确率': f"{results['accuracy']:.4f}",
            '对数损失': f"{results['log_loss']:.4f}",
            'MAE': f"{results['mae']:.4f}",
            'RMSE': f"{results['rmse']:.4f}",
            'KL散度': f"{results['kl_divergence']:.4f}"
        })
        
        # 打印每个类别的性能
        print(f"\n{model_type} 各类别精确率:")
        for class_name, precision in results['per_class_precision'].items():
            print(f"  {class_name}: {precision:.3f}")
    
    # 汇总结果
    results_df = pd.DataFrame(results_summary)
    print("\n\n模型性能对比:")
    print(results_df.to_string(index=False))
    
    return results_df


# 使用示例
if __name__ == "__main__":
    # 1. 比较不同模型
    # results_comparison = compare_models(transmod)
    
    # 2. 使用最佳模型（通常是XGBoost）
    best_model = EnhancedTransportModeModel(model_type='xgboost')
    (X_train, X_test, y_train_num, y_test_num, 
     y_probs_train, y_probs_test, y_train_str, y_test_str) = best_model.prepare_data(
        transmod, use_enhanced_features=True
    )
    best_model.train_with_optimization(X_train, y_train_num)
    results, predictions = best_model.evaluate_comprehensive(
        X_test, y_test_num, y_test_str, y_probs_test
    )
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

def predict_new_data(trained_model, data_to_pred):
    """
    使用训练好的模型预测新数据
    
    Parameters:
    -----------
    trained_model : EnhancedTransportModeModel
        已经训练好的模型实例
    df_imputed: pd.DataFrame
        需要预测的新数据
        
    Returns:
    --------
    pd.DataFrame : 包含原始数据和预测结果的DataFrame
    """
    
    # 1. 复制数据，避免修改原始数据
    data = data_to_pred.copy()
    
    # 2. 创建增强特征（使用模型中的create_features方法）
    if trained_model.enhanced_features is not None:
        # 使用模型的特征创建方法
        data_enhanced, _ = trained_model.create_features(data)
        features_to_use = trained_model.enhanced_features
    else:
        data_enhanced = data
        features_to_use = trained_model.feature_names
    
    # 3. 处理缺失值（使用训练时的imputer）
    if trained_model.imputer is not None:
        data_enhanced[features_to_use] = trained_model.imputer.transform(data_enhanced[features_to_use])
    else:
        # 如果模型没有保存imputer，创建一个新的
        imputer = SimpleImputer(strategy='median')
        data_enhanced[features_to_use] = imputer.fit_transform(data_enhanced[features_to_use])
    
    # 4. 提取特征
    X_new = data_enhanced[features_to_use].values
    
    # 5. 标准化（使用训练时的scaler）
    X_new_scaled = trained_model.scaler.transform(X_new)
    
    # 6. 进行预测
    # 预测类别
    y_pred_labels = trained_model.model.predict(X_new_scaled)
    # 预测概率
    y_pred_probs = trained_model.model.predict_proba(X_new_scaled)
    
    # 7. 将预测结果添加到原始数据中
    # 添加预测的类别（最可能的交通方式）
    y_pred_class_names = [trained_model.classes_[i] for i in y_pred_labels]
    data['predicted_mode'] = y_pred_class_names
    
    # 添加每种交通方式的概率
    data['car'] = y_pred_probs[:, 0]
    data['train'] = y_pred_probs[:, 1]
    data['plane'] = y_pred_probs[:, 2]
    
    # 8. 添加置信度（最高概率值）
    data['confidence'] = y_pred_probs.max(axis=1)
    
    return data


def save_predictions(predictions_df, output_path='predictions_2024.csv'):
    """
    保存预测结果到CSV文件
    
    Parameters:
    -----------
    predictions_df : pd.DataFrame
        包含预测结果的DataFrame
    output_path : str
        输出文件路径
    """
    # 选择要保存的列
    columns_to_save = ['date', 'origin', 'dest', 'predicted_mode', 
                       'car', 'train', 'plane', 'confidence']
    
    # 检查哪些列存在
    existing_columns = [col for col in columns_to_save if col in predictions_df.columns]
    
    # 保存结果
    predictions_df[existing_columns].to_csv(output_path, index=False)
    print(f"预测结果已保存到: {output_path}")


# 使用示例
if __name__ == "__main__":
    # 假设你已经有了训练好的模型 best_model
    # 以及需要预测的数据 data_intensity_2024
    predictions = predict_new_data(best_model, df_imputed)
    
    # 方法1：如果你已经有训练好的模型实例
    """
    # 进行预测
    predictions = predict_new_data(best_model, data_intensity_2024)
    
    # 查看前几行预测结果
    print("预测结果示例：")
    print(predictions[['origin', 'dest', 'predicted_mode', 'car', 'train', 'plane']].head())
    
    # 保存预测结果
    save_predictions(predictions)
    """
    
    # 方法2：如果需要重新加载模型（需要先保存模型）
    """
    import joblib
    
    # 保存模型
    joblib.dump(best_model, 'transport_mode_model.pkl')
    
    # 加载模型
    loaded_model = joblib.load('transport_mode_model.pkl')
    
    # 使用加载的模型进行预测
    predictions = predict_new_data(loaded_model, data_intensity_2024)
    """
    
    # 方法3：完整的预测流程
    """
    # 1. 创建并训练模型（使用你的训练数据）
    best_model = EnhancedTransportModeModel(model_type='xgboost')
    (X_train, X_test, y_train_num, y_test_num, 
     y_probs_train, y_probs_test, y_train_str, y_test_str) = best_model.prepare_data(
        transmod, use_enhanced_features=True
    )
    best_model.train_with_optimization(X_train, y_train_num)
    
    # 2. 使用训练好的模型预测新数据
    predictions = predict_new_data(best_model, data_intensity_2024)
    
    # 3. 查看预测结果
    print("\n预测结果统计：")
    print(predictions['predicted_mode'].value_counts())
    print("\n平均概率分布：")
    print(f"Car: {predictions['car'].mean():.3f}")
    print(f"Train: {predictions['train'].mean():.3f}")
    print(f"Plane: {predictions['plane'].mean():.3f}")
    
    # 4. 保存结果
    save_predictions(predictions)
    """
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import xgboost as xgb
from scipy.special import softmax
import warnings
warnings.filterwarnings('ignore')

# ========== 第一部分：城市距离管理器（保持不变） ==========
class CityDistanceManager:
    """管理城市间距离数据"""
    
    def __init__(self, distance_file_path=r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\city_dist.csv'):
        self.distance_file_path = distance_file_path
        self.distance_dict = {}
        self.load_distances()
    
    def load_distances(self):
        """加载城市距离数据"""
        try:
            df_dist = pd.read_csv(self.distance_file_path)
            for _, row in df_dist.iterrows():
                city1 = row['城市1']
                city2 = row['城市2']
                distance = row['距离(公里)']
                self.distance_dict[(city1, city2)] = distance
                self.distance_dict[(city2, city1)] = distance
            print(f"成功加载 {len(df_dist)} 条城市距离数据")
        except Exception as e:
            print(f"加载距离文件时出错: {e}")
            print("将使用默认距离值")
    
    def get_distance(self, origin, dest):
        """获取两个城市之间的距离"""
        return self.distance_dict.get((origin, dest), None)


# ========== 第二部分：交通模式占比回归模型 ==========
class TransportModeShareRegressor:
    """交通模式占比回归模型"""
    
    def __init__(self, model_type='xgboost', distance_file_path=r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\city_dist.csv', random_state=42):
        self.model_type = model_type
        self.models = {}  # 为每种交通方式存储一个模型
        self.scaler = StandardScaler()
        self.imputer = None
        self.random_state = random_state
        self.feature_names = None
        self.enhanced_features = None
        self.distance_manager = CityDistanceManager(distance_file_path)
        
        # 距离阈值设置
        self.PLANE_MIN_DISTANCE = 300
        self.TRAIN_MIN_DISTANCE = 50
        self.CAR_MAX_DISTANCE = 800
        self.TRAIN_OPTIMAL_RANGE = (100, 1000)
        
    def add_distance_features(self, df):
        """为数据添加距离特征"""
        data = df.copy()
        
        if 'origin' in data.columns and 'dest' in data.columns:
            data['distance'] = data.apply(
                lambda row: self.distance_manager.get_distance(row['origin'], row['dest']), 
                axis=1
            )
            
            median_distance = data['distance'].median() if data['distance'].notna().any() else 200
            data['distance'].fillna(median_distance, inplace=True)
            
            # 基础距离特征
            data['log_distance'] = np.log1p(data['distance'])
            data['distance_squared'] = data['distance'] ** 2
            data['sqrt_distance'] = np.sqrt(data['distance'])
            
            # 交通方式适合度特征（连续值）
            data['plane_suitability'] = 1 / (1 + np.exp(-(data['distance'] - self.PLANE_MIN_DISTANCE) / 50))
            
            train_center = (self.TRAIN_OPTIMAL_RANGE[0] + self.TRAIN_OPTIMAL_RANGE[1]) / 2
            train_width = (self.TRAIN_OPTIMAL_RANGE[1] - self.TRAIN_OPTIMAL_RANGE[0]) / 4
            data['train_suitability'] = np.exp(-0.5 * ((data['distance'] - train_center) / train_width) ** 2)
            
            data['car_suitability'] = 1 / (1 + np.exp((data['distance'] - self.CAR_MAX_DISTANCE / 2) / 50))
            
            # 距离区间特征
            data['is_short_distance'] = (data['distance'] < 100).astype(int)
            data['is_medium_distance'] = ((data['distance'] >= 100) & (data['distance'] < 500)).astype(int)
            data['is_long_distance'] = (data['distance'] >= 500).astype(int)
            data['is_ultra_long_distance'] = (data['distance'] >= 1000).astype(int)
            
            # 距离分段
            data['distance_0_50'] = (data['distance'] < 50).astype(int)
            data['distance_50_150'] = ((data['distance'] >= 50) & (data['distance'] < 150)).astype(int)
            data['distance_150_300'] = ((data['distance'] >= 150) & (data['distance'] < 300)).astype(int)
            data['distance_300_600'] = ((data['distance'] >= 300) & (data['distance'] < 600)).astype(int)
            data['distance_600_1000'] = ((data['distance'] >= 600) & (data['distance'] < 1000)).astype(int)
            data['distance_1000_plus'] = (data['distance'] >= 1000).astype(int)
            
            print(f"距离统计: 最小={data['distance'].min():.1f}km, "
                  f"最大={data['distance'].max():.1f}km, "
                  f"平均={data['distance'].mean():.1f}km")
        
        return data
    
    def create_features(self, df):
        """创建所有增强特征"""
        data = self.add_distance_features(df)
        
        # 基础设施对比特征
        data['AirFV_diff'] = data['AirFV_o'] - data['AirFV_d']
        data['RailFV_diff'] = data['RailFV_o'] - data['RailFV_d']
        data['RoadFV_diff'] = data['RoadFV_o'] - data['RoadFV_d']
        
        data['Air_vs_Rail_o'] = data['AirFV_o'] / (data['RailFV_o'] + 1e-6)
        data['Air_vs_Road_o'] = data['AirFV_o'] / (data['RoadFV_o'] + 1e-6)
        data['Rail_vs_Road_o'] = data['RailFV_o'] / (data['RoadFV_o'] + 1e-6)
        
        data['Air_vs_Rail_d'] = data['AirFV_d'] / (data['RailFV_d'] + 1e-6)
        data['Air_vs_Road_d'] = data['AirFV_d'] / (data['RoadFV_d'] + 1e-6)
        data['Rail_vs_Road_d'] = data['RailFV_d'] / (data['RoadFV_d'] + 1e-6)
        
        # 经济特征
        data['GDP_diff'] = data['GDP_o'] - data['GDP_d']
        data['GDP_ratio'] = data['GDP_o'] / (data['GDP_d'] + 1e-6)
        data['GDP_mean'] = (data['GDP_o'] + data['GDP_d']) / 2
        data['GDP_product'] = data['GDP_o'] * data['GDP_d']
        
        # 人口特征
        data['population_diff'] = data['population_o'] - data['population_d']
        data['population_ratio'] = data['population_o'] / (data['population_d'] + 1e-6)
        data['population_mean'] = (data['population_o'] + data['population_d']) / 2
        
        # 距离与基础设施交互特征
        if 'distance' in data.columns:
            # 针对每种交通方式的特定交互
            data['distance_x_air_infra'] = data['distance'] * (data['AirFV_o'] + data['AirFV_d']) / 2
            data['distance_x_rail_infra'] = data['distance'] * (data['RailFV_o'] + data['RailFV_d']) / 2
            data['distance_x_road_infra'] = data['distance'] * (data['RoadFV_o'] + data['RoadFV_d']) / 2
            
            # 距离与经济发展的交互
            data['distance_x_gdp'] = data['distance'] * data['GDP_mean']
            data['distance_x_population'] = data['distance'] * data['population_mean']
            
            # 特定距离范围与基础设施的交互
            data['short_dist_x_road'] = data['is_short_distance'] * (data['RoadFV_o'] + data['RoadFV_d']) / 2
            data['medium_dist_x_rail'] = data['is_medium_distance'] * (data['RailFV_o'] + data['RailFV_d']) / 2
            data['long_dist_x_air'] = data['is_long_distance'] * (data['AirFV_o'] + data['AirFV_d']) / 2
            
            # 适合度与基础设施的交互
            data['car_suit_x_road'] = data['car_suitability'] * (data['RoadFV_o'] + data['RoadFV_d']) / 2
            data['train_suit_x_rail'] = data['train_suitability'] * (data['RailFV_o'] + data['RailFV_d']) / 2
            data['plane_suit_x_air'] = data['plane_suitability'] * (data['AirFV_o'] + data['AirFV_d']) / 2
        
        # 天气条件
        data['weather_severity_o'] = (
            data['PREC_o'] * 0.4 +
            np.abs(data['temax_o'] - 20) * 0.3 +
            data['WDSP_o'] * 0.3
        )
        data['weather_severity_d'] = (
            data['PREC_d'] * 0.4 + 
            np.abs(data['temax_d'] - 20) * 0.3 +
            data['WDSP_d'] * 0.3
        )
        data['weather_diff'] = data['weather_severity_o'] - data['weather_severity_d']
        
        # 列出所有新特征
        new_features = [
            'distance', 'log_distance', 'distance_squared', 'sqrt_distance',
            'plane_suitability', 'train_suitability', 'car_suitability',
            'is_short_distance', 'is_medium_distance', 'is_long_distance', 'is_ultra_long_distance',
            'distance_0_50', 'distance_50_150', 'distance_150_300', 'distance_300_600', 
            'distance_600_1000', 'distance_1000_plus',
            'distance_x_air_infra', 'distance_x_rail_infra', 'distance_x_road_infra',
            'distance_x_gdp', 'distance_x_population',
            'short_dist_x_road', 'medium_dist_x_rail', 'long_dist_x_air',
            'car_suit_x_road', 'train_suit_x_rail', 'plane_suit_x_air',
            'AirFV_diff', 'RailFV_diff', 'RoadFV_diff',
            'Air_vs_Rail_o', 'Air_vs_Road_o', 'Rail_vs_Road_o',
            'Air_vs_Rail_d', 'Air_vs_Road_d', 'Rail_vs_Road_d',
            'GDP_diff', 'GDP_ratio', 'GDP_mean', 'GDP_product',
            'population_diff', 'population_ratio', 'population_mean',
            'weather_severity_o', 'weather_severity_d', 'weather_diff'
        ]
        
        return data, new_features
    
    def prepare_data(self, transmod, test_size=0.2, use_enhanced_features=True):
        """准备数据用于回归"""
        # 基础特征
        self.feature_names = [
            'AirFV_o', 'AirFV_d', 'AirPV_o', 'AirPV_d', 'GDP_o', 'GDP_d', 
            'NDVI_o', 'NDVI_d', 'population_o', 'population_d', 
            'RailFV_o', 'RailFV_d', 'RailPV_o', 'RailPV_d', 
            'RoadFV_o', 'RoadFV_d', 'RoadPV_o', 'RoadPV_d', 
            'PREC_o', 'PREC_d', 
            'temax_o', 'temax_d', 'WDSP_o', 'WDSP_d', 
            'is_workday?', 'is_weekend?', 'is_holiday?'
        ]
        
        # 创建增强特征
        if use_enhanced_features:
            data, new_features = self.create_features(transmod)
            self.enhanced_features = self.feature_names + new_features
            features_to_use = self.enhanced_features
        else:
            data = transmod.copy()
            features_to_use = self.feature_names
        
        # 处理缺失值
        self.imputer = SimpleImputer(strategy='median')
        data[features_to_use] = self.imputer.fit_transform(data[features_to_use])
        
        # 提取特征和目标（占比）
        X = data[features_to_use].values
        y_shares = data[['car', 'train', 'plane']].values
        
        # 确保占比和为1
        y_shares = y_shares / y_shares.sum(axis=1, keepdims=True)
        
        # 划分数据集
        X_train, X_test, y_train, y_test = train_test_split(
            X, y_shares, 
            test_size=test_size, 
            random_state=self.random_state
        )
        
        # 标准化
        X_train_scaled = self.scaler.fit_transform(X_train)
        X_test_scaled = self.scaler.transform(X_test)
        
        return X_train_scaled, X_test_scaled, y_train, y_test
    
    def train(self, X_train, y_train):
        """训练多输出回归模型"""
        if self.model_type == 'xgboost':
            # 为每种交通方式训练一个模型
            modes = ['car', 'train', 'plane']
            
            for i, mode in enumerate(modes):
                print(f"训练{mode}占比预测模型...")
                
                # 创建XGBoost回归器
                xgb_model = xgb.XGBRegressor(
                    n_estimators=300,
                    max_depth=6,
                    learning_rate=0.03,
                    subsample=0.8,
                    colsample_bytree=0.8,
                    objective='reg:squarederror',
                    random_state=self.random_state,
                    n_jobs=-1
                )
                
                # 训练模型
                xgb_model.fit(X_train, y_train[:, i])
                self.models[mode] = xgb_model
                
                # 计算训练集R²
                train_pred = xgb_model.predict(X_train)
                train_r2 = r2_score(y_train[:, i], train_pred)
                print(f"  {mode}训练集R²: {train_r2:.3f}")
        else:
            print(f"模型类型 {self.model_type} 暂未实现")
    
    def predict(self, X):
        """预测占比并确保和为1"""
        predictions = np.zeros((X.shape[0], 3))
        
        # 获取每种交通方式的预测
        predictions[:, 0] = self.models['car'].predict(X)
        predictions[:, 1] = self.models['train'].predict(X)
        predictions[:, 2] = self.models['plane'].predict(X)
        
        # 确保预测值非负
        predictions = np.maximum(predictions, 0)
        
        # 归一化使和为1
        row_sums = predictions.sum(axis=1, keepdims=True)
        predictions = predictions / (row_sums + 1e-8)
        
        return predictions
    
    def evaluate(self, X_test, y_test):
        """评估模型性能"""
        y_pred = self.predict(X_test)
        
        print("\n=== 模型评估结果 ===")
        
        # 计算每种交通方式的指标
        modes = ['car', 'train', 'plane']
        for i, mode in enumerate(modes):
            mse = mean_squared_error(y_test[:, i], y_pred[:, i])
            mae = mean_absolute_error(y_test[:, i], y_pred[:, i])
            r2 = r2_score(y_test[:, i], y_pred[:, i])
            
            print(f"\n{mode}占比预测:")
            print(f"  MSE: {mse:.4f}")
            print(f"  MAE: {mae:.4f}")
            print(f"  R²: {r2:.3f}")
            
            # 预测值与真实值的对比
            print(f"  真实占比范围: [{y_test[:, i].min():.2f}, {y_test[:, i].max():.2f}]")
            print(f"  预测占比范围: [{y_pred[:, i].min():.2f}, {y_pred[:, i].max():.2f}]")
        
        # 总体指标
        total_mae = mean_absolute_error(y_test, y_pred)
        print(f"\n总体平均绝对误差: {total_mae:.4f}")
        
        return y_pred


# ========== 第三部分：预测函数 ==========
def predict_mode_shares(trained_model, data_to_pred):
    """使用训练好的模型预测交通模式占比"""
    data = data_to_pred.copy()
    
    # 创建增强特征
    if trained_model.enhanced_features is not None:
        data_enhanced, _ = trained_model.create_features(data)
        features_to_use = trained_model.enhanced_features
    else:
        data_enhanced = data
        features_to_use = trained_model.feature_names
    
    # 处理缺失值
    if trained_model.imputer is not None:
        data_enhanced[features_to_use] = trained_model.imputer.transform(data_enhanced[features_to_use])
    
    # 提取特征
    X_new = data_enhanced[features_to_use].values
    
    # 标准化
    X_new_scaled = trained_model.scaler.transform(X_new)
    
    # 预测占比
    y_pred_shares = trained_model.predict(X_new_scaled)
    
    # 添加结果到数据框
    data['car_share'] = y_pred_shares[:, 0]
    data['train_share'] = y_pred_shares[:, 1]
    data['plane_share'] = y_pred_shares[:, 2]
    
    # 识别主要交通方式
    max_indices = y_pred_shares.argmax(axis=1)
    modes = ['car', 'train', 'plane']
    data['dominant_mode'] = [modes[i] for i in max_indices]
    data['max_share'] = y_pred_shares.max(axis=1)
    
    # 添加距离信息
    if 'distance' in data_enhanced.columns:
        data['distance'] = data_enhanced['distance']
    
    return data


def analyze_beijing_shanghai_predictions(predictions_df, training_df):
    """分析北京-上海的预测结果并与训练数据对比"""
    # 筛选北京-上海的预测结果
    bj_sh_pred = predictions_df[
        (predictions_df['origin'] == '北京') & 
        (predictions_df['dest'] == '上海')
    ].copy()
    
    # 筛选北京-上海的训练数据
    bj_sh_train = training_df[
        (training_df['origin'] == '北京') & 
        (training_df['dest'] == '上海')
    ].copy()
    
    if len(bj_sh_pred) > 0 and len(bj_sh_train) > 0:
        print("=== 北京-上海 预测结果分析 ===\n")
        
        # 训练数据统计
        print("训练数据中的占比分布:")
        print(f"  汽车: 平均={bj_sh_train['car'].mean():.1%}, "
              f"范围=[{bj_sh_train['car'].min():.1%}, {bj_sh_train['car'].max():.1%}]")
        print(f"  火车: 平均={bj_sh_train['train'].mean():.1%}, "
              f"范围=[{bj_sh_train['train'].min():.1%}, {bj_sh_train['train'].max():.1%}]")
        print(f"  飞机: 平均={bj_sh_train['plane'].mean():.1%}, "
              f"范围=[{bj_sh_train['plane'].min():.1%}, {bj_sh_train['plane'].max():.1%}]")
        
        # 预测结果统计
        print("\n预测结果中的占比分布:")
        print(f"  汽车: 平均={bj_sh_pred['car_share'].mean():.1%}, "
              f"范围=[{bj_sh_pred['car_share'].min():.1%}, {bj_sh_pred['car_share'].max():.1%}]")
        print(f"  火车: 平均={bj_sh_pred['train_share'].mean():.1%}, "
              f"范围=[{bj_sh_pred['train_share'].min():.1%}, {bj_sh_pred['train_share'].max():.1%}]")
        print(f"  飞机: 平均={bj_sh_pred['plane_share'].mean():.1%}, "
              f"范围=[{bj_sh_pred['plane_share'].min():.1%}, {bj_sh_pred['plane_share'].max():.1%}]")
        
        # 显示几个具体样本
        print("\n预测样本示例:")
        sample_cols = ['date', 'car_share', 'train_share', 'plane_share', 'distance'] if 'date' in bj_sh_pred.columns else ['car_share', 'train_share', 'plane_share', 'distance']
        print(bj_sh_pred[sample_cols].head())

def save_model(self, model_path="transport_model"):
    """保存训练好的模型"""
    # 创建保存目录
    if not os.path.exists(model_path):
        os.makedirs(model_path)
        
    # 保存模型组件
    model_data = {
        'models': self.models,
        'scaler': self.scaler,
        'imputer': self.imputer,
        'feature_names': self.feature_names,
        'enhanced_features': self.enhanced_features,
        'model_type': self.model_type,
        'random_state': self.random_state,
        'distance_manager': self.distance_manager,
        'train_time': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
        
     # 保存到pickle文件
    with open(f"{model_path}/model_components.pkl", 'wb') as f:
        pickle.dump(model_data, f)
        
    # 分别保存每个XGBoost模型（可选，用于兼容性）
    for mode, model in self.models.items():
        model.save_model(f"{model_path}/{mode}_model.json")
        
    print(f"模型已保存到: {model_path}")
    print(f"保存时间: {model_data['train_time']}")
    
def load_model(self, model_path="transport_model"):
    """加载保存的模型"""
    try:
        # 加载模型组件
        with open(f"{model_path}/model_components.pkl", 'rb') as f:
            model_data = pickle.load(f)
            
        # 恢复模型属性
        self.models = model_data['models']
        self.scaler = model_data['scaler']
        self.imputer = model_data['imputer']
        self.feature_names = model_data['feature_names']
        self.enhanced_features = model_data['enhanced_features']
        self.model_type = model_data['model_type']
        self.random_state = model_data['random_state']
        self.distance_manager = model_data['distance_manager']
            
        print(f"模型加载成功!")
        print(f"模型训练时间: {model_data.get('train_time', '未知')}")
        print(f"包含交通方式: {list(self.models.keys())}")
        return True
            
    except FileNotFoundError:
        print(f"模型文件不存在: {model_path}")
        return False
    except Exception as e:
        print(f"加载模型时出错: {e}")
        return False


# ========== 使用示例 ==========
# 1. 创建回归模型
print("创建交通模式占比回归模型...")
share_model = TransportModeShareRegressor(
    model_type='xgboost',
    distance_file_path=r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\city_dist.csv'
)

# 2. 准备数据
print("\n准备训练数据...")
X_train, X_test, y_train, y_test = share_model.prepare_data(
    transmod,  # 你的训练数据
    use_enhanced_features=True
)

# 3. 训练模型
print("\n训练回归模型...")
share_model.train(X_train, y_train)

# 4. 评估模型
print("\n评估模型性能...")
y_pred_test = share_model.evaluate(X_test, y_test)

# 5. 预测2024年数据
print("\n预测2024年数据...")
predictions_2024 = predict_mode_shares(share_model, df_imputed)

# 6. 分析北京-上海的结果
analyze_beijing_shanghai_predictions(predictions_2024, transmod)

# 7. 查看更多城市对的结果
print("\n=== 其他城市对示例 ===")

# 短距离示例
short_routes = predictions_2024[predictions_2024['distance'] < 100].head(5)
if len(short_routes) > 0:
    print("\n短距离路线(<100km):")
    print(short_routes[['origin', 'dest', 'distance', 'car_share', 'train_share', 'plane_share']])

# 中距离示例  
medium_routes = predictions_2024[(predictions_2024['distance'] >= 300) & (predictions_2024['distance'] < 600)].head(5)
if len(medium_routes) > 0:
    print("\n中距离路线(300-600km):")
    print(medium_routes[['origin', 'dest', 'distance', 'car_share', 'train_share', 'plane_share']])

# 长距离示例
long_routes = predictions_2024[predictions_2024['distance'] > 1000].head(5)
if len(long_routes) > 0:
    print("\n长距离路线(>1000km):")
    print(long_routes[['origin', 'dest', 'distance', 'car_share', 'train_share', 'plane_share']])

# 8. 保存结果
print("\n保存预测结果...")
predictions_2024.to_csv('predictions_2024_shares_regression.csv', index=False)
print("结果已保存到 predictions_2024_shares_regression.csv")

# 9. 可选：特征重要性分析
# print("\n=== 特征重要性分析 ===")
# for mode in ['car', 'train', 'plane']:
#     print(f"\n{mode}占比预测的重要特征:")
#     feature_importance = share_model.models[mode].feature_importances_
#     feature_names = share_model.enhanced_features if share_model.enhanced_features else share_model.feature_names
    
#     # 获取前10个最重要的特征
#     indices = np.argsort(feature_importance)[-10:][::-1]
#     for idx in indices[:5]:  # 只显示前5个
#         print(f"  {feature_names[idx]}: {feature_importance[idx]:.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import xgboost as xgb
from scipy.special import softmax
import warnings
import pickle
import os
from datetime import datetime
import matplotlib.pyplot as plt

warnings.filterwarnings('ignore')

# ========== 第一部分：城市距离管理器 ==========
class CityDistanceManager:
    """管理城市间距离数据"""
    
    def __init__(self, distance_file_path=r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\city_dist.csv'):
        self.distance_file_path = distance_file_path
        self.distance_dict = {}
        self.load_distances()
    
    def load_distances(self):
        """加载城市距离数据"""
        try:
            df_dist = pd.read_csv(self.distance_file_path)
            for _, row in df_dist.iterrows():
                city1 = row['城市1']
                city2 = row['城市2']
                distance = row['距离(公里)']
                self.distance_dict[(city1, city2)] = distance
                self.distance_dict[(city2, city1)] = distance
            print(f"成功加载 {len(df_dist)} 条城市距离数据")
        except Exception as e:
            print(f"加载距离文件时出错: {e}")
            print("将使用默认距离值")
    
    def get_distance(self, origin, dest):
        """获取两个城市之间的距离"""
        return self.distance_dict.get((origin, dest), None)


# ========== 第二部分：交通模式占比回归模型 ==========
class TransportModeShareRegressor:
    """交通模式占比回归模型"""
    
    def __init__(self, model_type='xgboost', distance_file_path=r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\city_dist.csv', random_state=42):
        self.model_type = model_type
        self.models = {}  # 为每种交通方式存储一个模型
        self.scaler = StandardScaler()
        self.imputer = None
        self.random_state = random_state
        self.feature_names = None
        self.enhanced_features = None
        self.distance_manager = CityDistanceManager(distance_file_path)
        
        # 距离阈值设置
        self.PLANE_MIN_DISTANCE = 300
        self.TRAIN_MIN_DISTANCE = 50
        self.CAR_MAX_DISTANCE = 800
        self.TRAIN_OPTIMAL_RANGE = (100, 1000)
        
    def add_distance_features(self, df):
        """为数据添加距离特征"""
        data = df.copy()
        
        if 'origin' in data.columns and 'dest' in data.columns:
            data['distance'] = data.apply(
                lambda row: self.distance_manager.get_distance(row['origin'], row['dest']), 
                axis=1
            )
            
            median_distance = data['distance'].median() if data['distance'].notna().any() else 200
            data['distance'].fillna(median_distance, inplace=True)
            
            # 基础距离特征
            data['log_distance'] = np.log1p(data['distance'])
            data['distance_squared'] = data['distance'] ** 2
            data['sqrt_distance'] = np.sqrt(data['distance'])
            
            # 交通方式适合度特征（连续值）
            data['plane_suitability'] = 1 / (1 + np.exp(-(data['distance'] - self.PLANE_MIN_DISTANCE) / 50))
            
            train_center = (self.TRAIN_OPTIMAL_RANGE[0] + self.TRAIN_OPTIMAL_RANGE[1]) / 2
            train_width = (self.TRAIN_OPTIMAL_RANGE[1] - self.TRAIN_OPTIMAL_RANGE[0]) / 4
            data['train_suitability'] = np.exp(-0.5 * ((data['distance'] - train_center) / train_width) ** 2)
            
            data['car_suitability'] = 1 / (1 + np.exp((data['distance'] - self.CAR_MAX_DISTANCE / 2) / 50))
            
            # 距离区间特征
            data['is_short_distance'] = (data['distance'] < 100).astype(int)
            data['is_medium_distance'] = ((data['distance'] >= 100) & (data['distance'] < 500)).astype(int)
            data['is_long_distance'] = (data['distance'] >= 500).astype(int)
            data['is_ultra_long_distance'] = (data['distance'] >= 1000).astype(int)
            
            # 距离分段
            data['distance_0_50'] = (data['distance'] < 50).astype(int)
            data['distance_50_150'] = ((data['distance'] >= 50) & (data['distance'] < 150)).astype(int)
            data['distance_150_300'] = ((data['distance'] >= 150) & (data['distance'] < 300)).astype(int)
            data['distance_300_600'] = ((data['distance'] >= 300) & (data['distance'] < 600)).astype(int)
            data['distance_600_1000'] = ((data['distance'] >= 600) & (data['distance'] < 1000)).astype(int)
            data['distance_1000_plus'] = (data['distance'] >= 1000).astype(int)
            
            print(f"距离统计: 最小={data['distance'].min():.1f}km, "
                  f"最大={data['distance'].max():.1f}km, "
                  f"平均={data['distance'].mean():.1f}km")
        
        return data
    
    def create_features(self, df):
        """创建所有增强特征"""
        data = self.add_distance_features(df)
        
        # 基础设施对比特征
        data['AirFV_diff'] = data['AirFV_o'] - data['AirFV_d']
        data['RailFV_diff'] = data['RailFV_o'] - data['RailFV_d']
        data['RoadFV_diff'] = data['RoadFV_o'] - data['RoadFV_d']
        
        data['Air_vs_Rail_o'] = data['AirFV_o'] / (data['RailFV_o'] + 1e-6)
        data['Air_vs_Road_o'] = data['AirFV_o'] / (data['RoadFV_o'] + 1e-6)
        data['Rail_vs_Road_o'] = data['RailFV_o'] / (data['RoadFV_o'] + 1e-6)
        
        data['Air_vs_Rail_d'] = data['AirFV_d'] / (data['RailFV_d'] + 1e-6)
        data['Air_vs_Road_d'] = data['AirFV_d'] / (data['RoadFV_d'] + 1e-6)
        data['Rail_vs_Road_d'] = data['RailFV_d'] / (data['RoadFV_d'] + 1e-6)
        
        # 经济特征
        data['GDP_diff'] = data['GDP_o'] - data['GDP_d']
        data['GDP_ratio'] = data['GDP_o'] / (data['GDP_d'] + 1e-6)
        data['GDP_mean'] = (data['GDP_o'] + data['GDP_d']) / 2
        data['GDP_product'] = data['GDP_o'] * data['GDP_d']
        
        # 人口特征
        data['population_diff'] = data['population_o'] - data['population_d']
        data['population_ratio'] = data['population_o'] / (data['population_d'] + 1e-6)
        data['population_mean'] = (data['population_o'] + data['population_d']) / 2
        
        # 距离与基础设施交互特征
        if 'distance' in data.columns:
            # 针对每种交通方式的特定交互
            data['distance_x_air_infra'] = data['distance'] * (data['AirFV_o'] + data['AirFV_d']) / 2
            data['distance_x_rail_infra'] = data['distance'] * (data['RailFV_o'] + data['RailFV_d']) / 2
            data['distance_x_road_infra'] = data['distance'] * (data['RoadFV_o'] + data['RoadFV_d']) / 2
            
            # 距离与经济发展的交互
            data['distance_x_gdp'] = data['distance'] * data['GDP_mean']
            data['distance_x_population'] = data['distance'] * data['population_mean']
            
            # 特定距离范围与基础设施的交互
            data['short_dist_x_road'] = data['is_short_distance'] * (data['RoadFV_o'] + data['RoadFV_d']) / 2
            data['medium_dist_x_rail'] = data['is_medium_distance'] * (data['RailFV_o'] + data['RailFV_d']) / 2
            data['long_dist_x_air'] = data['is_long_distance'] * (data['AirFV_o'] + data['AirFV_d']) / 2
            
            # 适合度与基础设施的交互
            data['car_suit_x_road'] = data['car_suitability'] * (data['RoadFV_o'] + data['RoadFV_d']) / 2
            data['train_suit_x_rail'] = data['train_suitability'] * (data['RailFV_o'] + data['RailFV_d']) / 2
            data['plane_suit_x_air'] = data['plane_suitability'] * (data['AirFV_o'] + data['AirFV_d']) / 2
        
        # 天气条件
        data['weather_severity_o'] = (
            data['PREC_o'] * 0.4 +
            np.abs(data['temax_o'] - 20) * 0.3 +
            data['WDSP_o'] * 0.3
        )
        data['weather_severity_d'] = (
            data['PREC_d'] * 0.4 + 
            np.abs(data['temax_d'] - 20) * 0.3 +
            data['WDSP_d'] * 0.3
        )
        data['weather_diff'] = data['weather_severity_o'] - data['weather_severity_d']
        
        # 列出所有新特征
        new_features = [
            'distance', 'log_distance', 'distance_squared', 'sqrt_distance',
            'plane_suitability', 'train_suitability', 'car_suitability',
            'is_short_distance', 'is_medium_distance', 'is_long_distance', 'is_ultra_long_distance',
            'distance_0_50', 'distance_50_150', 'distance_150_300', 'distance_300_600', 
            'distance_600_1000', 'distance_1000_plus',
            'distance_x_air_infra', 'distance_x_rail_infra', 'distance_x_road_infra',
            'distance_x_gdp', 'distance_x_population',
            'short_dist_x_road', 'medium_dist_x_rail', 'long_dist_x_air',
            'car_suit_x_road', 'train_suit_x_rail', 'plane_suit_x_air',
            'AirFV_diff', 'RailFV_diff', 'RoadFV_diff',
            'Air_vs_Rail_o', 'Air_vs_Road_o', 'Rail_vs_Road_o',
            'Air_vs_Rail_d', 'Air_vs_Road_d', 'Rail_vs_Road_d',
            'GDP_diff', 'GDP_ratio', 'GDP_mean', 'GDP_product',
            'population_diff', 'population_ratio', 'population_mean',
            'weather_severity_o', 'weather_severity_d', 'weather_diff'
        ]
        
        return data, new_features
    
    def prepare_data(self, transmod, test_size=0.2, use_enhanced_features=True):
        """准备数据用于回归"""
        # 基础特征
        self.feature_names = [
            'AirFV_o', 'AirFV_d', 'AirPV_o', 'AirPV_d', 'GDP_o', 'GDP_d', 
            'NDVI_o', 'NDVI_d', 'population_o', 'population_d', 
            'RailFV_o', 'RailFV_d', 'RailPV_o', 'RailPV_d', 
            'RoadFV_o', 'RoadFV_d', 'RoadPV_o', 'RoadPV_d', 
            'PREC_o', 'PREC_d', 
            'temax_o', 'temax_d', 'WDSP_o', 'WDSP_d', 
            'is_workday?', 'is_weekend?', 'is_holiday?'
        ]
        
        # 创建增强特征
        if use_enhanced_features:
            data, new_features = self.create_features(transmod)
            self.enhanced_features = self.feature_names + new_features
            features_to_use = self.enhanced_features
        else:
            data = transmod.copy()
            features_to_use = self.feature_names
        
        # 处理缺失值
        self.imputer = SimpleImputer(strategy='median')
        data[features_to_use] = self.imputer.fit_transform(data[features_to_use])
        
        # 提取特征和目标（占比）
        X = data[features_to_use].values
        y_shares = data[['car', 'train', 'plane']].values
        
        # 确保占比和为1
        y_shares = y_shares / y_shares.sum(axis=1, keepdims=True)
        
        # 划分数据集
        X_train, X_test, y_train, y_test = train_test_split(
            X, y_shares, 
            test_size=test_size, 
            random_state=self.random_state
        )
        
        # 标准化
        X_train_scaled = self.scaler.fit_transform(X_train)
        X_test_scaled = self.scaler.transform(X_test)
        
        return X_train_scaled, X_test_scaled, y_train, y_test
    
    def train(self, X_train, y_train):
        """训练多输出回归模型"""
        if self.model_type == 'xgboost':
            # 为每种交通方式训练一个模型
            modes = ['car', 'train', 'plane']
            
            for i, mode in enumerate(modes):
                print(f"训练{mode}占比预测模型...")
                
                # 创建XGBoost回归器
                xgb_model = xgb.XGBRegressor(
                    n_estimators=300,
                    max_depth=6,
                    learning_rate=0.03,
                    subsample=0.8,
                    colsample_bytree=0.8,
                    objective='reg:squarederror',
                    random_state=self.random_state,
                    n_jobs=-1
                )
                
                # 训练模型
                xgb_model.fit(X_train, y_train[:, i])
                self.models[mode] = xgb_model
                
                # 计算训练集R²
                train_pred = xgb_model.predict(X_train)
                train_r2 = r2_score(y_train[:, i], train_pred)
                print(f"  {mode}训练集R²: {train_r2:.3f}")
        else:
            print(f"模型类型 {self.model_type} 暂未实现")
    
    def predict(self, X):
        """预测占比并确保和为1"""
        predictions = np.zeros((X.shape[0], 3))
        
        # 获取每种交通方式的预测
        predictions[:, 0] = self.models['car'].predict(X)
        predictions[:, 1] = self.models['train'].predict(X)
        predictions[:, 2] = self.models['plane'].predict(X)
        
        # 确保预测值非负
        predictions = np.maximum(predictions, 0)
        
        # 归一化使和为1
        row_sums = predictions.sum(axis=1, keepdims=True)
        predictions = predictions / (row_sums + 1e-8)
        
        return predictions
    
    def evaluate(self, X_test, y_test):
        """评估模型性能"""
        y_pred = self.predict(X_test)
        
        print("\n=== 模型评估结果 ===")
        
        # 计算每种交通方式的指标
        modes = ['car', 'train', 'plane']
        for i, mode in enumerate(modes):
            mse = mean_squared_error(y_test[:, i], y_pred[:, i])
            mae = mean_absolute_error(y_test[:, i], y_pred[:, i])
            r2 = r2_score(y_test[:, i], y_pred[:, i])
            
            print(f"\n{mode}占比预测:")
            print(f"  MSE: {mse:.4f}")
            print(f"  MAE: {mae:.4f}")
            print(f"  R²: {r2:.3f}")
            
            # 预测值与真实值的对比
            print(f"  真实占比范围: [{y_test[:, i].min():.2f}, {y_test[:, i].max():.2f}]")
            print(f"  预测占比范围: [{y_pred[:, i].min():.2f}, {y_pred[:, i].max():.2f}]")
        
        # 总体指标
        total_mae = mean_absolute_error(y_test, y_pred)
        print(f"\n总体平均绝对误差: {total_mae:.4f}")
        
        return y_pred

    def save_model(self, model_path="transport_model"):
        """保存训练好的模型"""
        # 创建保存目录
        if not os.path.exists(model_path):
            os.makedirs(model_path)
        
        # 保存模型组件
        model_data = {
            'models': self.models,
            'scaler': self.scaler,
            'imputer': self.imputer,
            'feature_names': self.feature_names,
            'enhanced_features': self.enhanced_features,
            'model_type': self.model_type,
            'random_state': self.random_state,
            'distance_manager': self.distance_manager,
            'train_time': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        }
        
        # 保存到pickle文件
        with open(f"{model_path}/model_components.pkl", 'wb') as f:
            pickle.dump(model_data, f)
        
        # 分别保存每个XGBoost模型（可选，用于兼容性）
        for mode, model in self.models.items():
            model.save_model(f"{model_path}/{mode}_model.json")
        
        print(f"模型已保存到: {model_path}")
        print(f"保存时间: {model_data['train_time']}")
    
    def load_model(self, model_path="transport_model"):
        """加载保存的模型"""
        try:
            # 加载模型组件
            with open(f"{model_path}/model_components.pkl", 'rb') as f:
                model_data = pickle.load(f)
            
            # 恢复模型属性
            self.models = model_data['models']
            self.scaler = model_data['scaler']
            self.imputer = model_data['imputer']
            self.feature_names = model_data['feature_names']
            self.enhanced_features = model_data['enhanced_features']
            self.model_type = model_data['model_type']
            self.random_state = model_data['random_state']
            self.distance_manager = model_data['distance_manager']
            
            print(f"模型加载成功!")
            print(f"模型训练时间: {model_data.get('train_time', '未知')}")
            print(f"包含交通方式: {list(self.models.keys())}")
            return True
            
        except FileNotFoundError:
            print(f"模型文件不存在: {model_path}")
            return False
        except Exception as e:
            print(f"加载模型时出错: {e}")
            return False


# ========== 第三部分：预测函数 ==========
def predict_mode_shares(trained_model, data_to_pred):
    """使用训练好的模型预测交通模式占比"""
    data = data_to_pred.copy()
    
    # 创建增强特征
    if trained_model.enhanced_features is not None:
        data_enhanced, _ = trained_model.create_features(data)
        features_to_use = trained_model.enhanced_features
    else:
        data_enhanced = data
        features_to_use = trained_model.feature_names
    
    # 处理缺失值
    if trained_model.imputer is not None:
        data_enhanced[features_to_use] = trained_model.imputer.transform(data_enhanced[features_to_use])
    
    # 提取特征
    X_new = data_enhanced[features_to_use].values
    
    # 标准化
    X_new_scaled = trained_model.scaler.transform(X_new)
    
    # 预测占比
    y_pred_shares = trained_model.predict(X_new_scaled)
    
    # 添加结果到数据框
    data['car_share'] = y_pred_shares[:, 0]
    data['train_share'] = y_pred_shares[:, 1]
    data['plane_share'] = y_pred_shares[:, 2]
    
    # 识别主要交通方式
    max_indices = y_pred_shares.argmax(axis=1)
    modes = ['car', 'train', 'plane']
    data['dominant_mode'] = [modes[i] for i in max_indices]
    data['max_share'] = y_pred_shares.max(axis=1)
    
    # 添加距离信息
    if 'distance' in data_enhanced.columns:
        data['distance'] = data_enhanced['distance']
    
    return data


def analyze_beijing_shanghai_predictions(predictions_df, training_df):
    """分析北京-上海的预测结果并与训练数据对比"""
    # 筛选北京-上海的预测结果
    bj_sh_pred = predictions_df[
        (predictions_df['origin'] == '北京') & 
        (predictions_df['dest'] == '上海')
    ].copy()
    
    # 筛选北京-上海的训练数据
    bj_sh_train = training_df[
        (training_df['origin'] == '北京') & 
        (training_df['dest'] == '上海')
    ].copy()
    
    if len(bj_sh_pred) > 0 and len(bj_sh_train) > 0:
        print("=== 北京-上海 预测结果分析 ===\n")
        
        # 训练数据统计
        print("训练数据中的占比分布:")
        print(f"  汽车: 平均={bj_sh_train['car'].mean():.1%}, "
              f"范围=[{bj_sh_train['car'].min():.1%}, {bj_sh_train['car'].max():.1%}]")
        print(f"  火车: 平均={bj_sh_train['train'].mean():.1%}, "
              f"范围=[{bj_sh_train['train'].min():.1%}, {bj_sh_train['train'].max():.1%}]")
        print(f"  飞机: 平均={bj_sh_train['plane'].mean():.1%}, "
              f"范围=[{bj_sh_train['plane'].min():.1%}, {bj_sh_train['plane'].max():.1%}]")
        
        # 预测结果统计
        print("\n预测结果中的占比分布:")
        print(f"  汽车: 平均={bj_sh_pred['car_share'].mean():.1%}, "
              f"范围=[{bj_sh_pred['car_share'].min():.1%}, {bj_sh_pred['car_share'].max():.1%}]")
        print(f"  火车: 平均={bj_sh_pred['train_share'].mean():.1%}, "
              f"范围=[{bj_sh_pred['train_share'].min():.1%}, {bj_sh_pred['train_share'].max():.1%}]")
        print(f"  飞机: 平均={bj_sh_pred['plane_share'].mean():.1%}, "
              f"范围=[{bj_sh_pred['plane_share'].min():.1%}, {bj_sh_pred['plane_share'].max():.1%}]")
        
        # 显示几个具体样本
        print("\n预测样本示例:")
        sample_cols = ['date', 'car_share', 'train_share', 'plane_share', 'distance'] if 'date' in bj_sh_pred.columns else ['car_share', 'train_share', 'plane_share', 'distance']
        print(bj_sh_pred[sample_cols].head())


# ========== 第四部分：SHAP分析功能 ==========
def install_shap_if_needed():
    """检查并安装SHAP库"""
    try:
        import shap
        print("SHAP库已安装")
        return True
    except ImportError:
        print("正在安装SHAP库...")
        import subprocess
        import sys
        try:
            subprocess.check_call([sys.executable, "-m", "pip", "install", "shap"])
            import shap
            print("SHAP库安装成功!")
            return True
        except:
            print("SHAP库安装失败，请手动安装: pip install shap")
            return False

def comprehensive_shap_analysis(model, X_test, y_test, feature_names, sample_size=1000):
    """完整的SHAP分析"""
    
    if not install_shap_if_needed():
        return
    
    import shap
    
    # 设置中文字体（如果需要显示中文）
    plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
    plt.rcParams['axes.unicode_minus'] = False
    
    # 限制样本数量以提高速度
    n_samples = min(sample_size, len(X_test))
    X_sample = X_test[:n_samples]
    y_sample = y_test[:n_samples]
    
    print(f"使用 {n_samples} 个样本进行SHAP分析...")
    
    modes = ['car', 'train', 'plane']
    mode_names = {'car': '汽车', 'train': '火车', 'plane': '飞机'}
    
    # 为每种交通方式进行分析
    for i, mode in enumerate(modes):
        print(f"\n{'='*50}")
        print(f"分析 {mode_names[mode]} 占比预测模型")
        print(f"{'='*50}")
        
        # 创建SHAP解释器
        explainer = shap.TreeExplainer(model.models[mode])
        
        # 计算SHAP值
        print("正在计算SHAP值...")
        shap_values = explainer.shap_values(X_sample)
        
        # 1. 特征重要性分析
        print(f"\n1. {mode_names[mode]} 的特征重要性排名:")
        feature_importance = np.abs(shap_values).mean(0)
        importance_df = pd.DataFrame({
            'feature': feature_names,
            'importance': feature_importance
        }).sort_values('importance', ascending=False)
        
        print("前10个最重要的特征:")
        for idx, (_, row) in enumerate(importance_df.head(10).iterrows()):
            print(f"  {idx+1:2d}. {row['feature']:<25} 重要性: {row['importance']:.4f}")
        
        # 2. 温度特征的具体影响
        print(f"\n2. 温度特征对 {mode_names[mode]} 的影响:")
        temp_features = [(idx, name) for idx, name in enumerate(feature_names) 
                        if 'temax' in name.lower()]
        
        for idx, feature_name in temp_features:
            shap_mean = shap_values[:, idx].mean()
            shap_std = shap_values[:, idx].std()
            importance = feature_importance[idx]
            
            direction = "正向提升" if shap_mean > 0 else "负向降低"
            strength = "强" if importance > 0.01 else "中等" if importance > 0.005 else "弱"
            
            print(f"  {feature_name}:")
            print(f"    平均SHAP值: {shap_mean:+.4f} ({direction})")
            print(f"    SHAP标准差: {shap_std:.4f}")
            print(f"    特征重要性: {importance:.4f} ({strength}影响)")
            print(f"    排名: 第{importance_df[importance_df['feature']==feature_name].index[0]+1}位")
        
        # 3. 绘制SHAP图
        plt.figure(figsize=(15, 10))
        
        # 3.1 特征重要性条形图
        plt.subplot(2, 2, 1)
        top_features = importance_df.head(15)
        plt.barh(range(len(top_features)), top_features['importance'])
        plt.yticks(range(len(top_features)), top_features['feature'])
        plt.xlabel('平均绝对SHAP值')
        plt.title(f'{mode_names[mode]} - 特征重要性排名')
        plt.gca().invert_yaxis()
        
        # 3.2 SHAP汇总图
        plt.subplot(2, 2, 2)
        try:
            shap.summary_plot(shap_values, X_sample, feature_names=feature_names, 
                             max_display=15, show=False)
            plt.title(f'{mode_names[mode]} - SHAP汇总图')
        except:
            plt.text(0.5, 0.5, 'SHAP汇总图生成失败', ha='center', va='center')
            plt.title(f'{mode_names[mode]} - SHAP汇总图 (生成失败)')
        
        # 3.3 温度特征的SHAP值分布
        plt.subplot(2, 2, 3)
        if temp_features:
            for idx, feature_name in temp_features:
                plt.hist(shap_values[:, idx], alpha=0.6, label=feature_name.replace('temax_', ''), bins=30)
            plt.xlabel('SHAP值')
            plt.ylabel('频次')
            plt.title(f'{mode_names[mode]} - 温度特征SHAP值分布')
            plt.legend()
            plt.axvline(x=0, color='red', linestyle='--', alpha=0.7)
        
        # 3.4 预测值 vs 实际值
        plt.subplot(2, 2, 4)
        y_pred_sample = model.models[mode].predict(X_sample)
        plt.scatter(y_sample[:, i], y_pred_sample, alpha=0.6)
        plt.plot([y_sample[:, i].min(), y_sample[:, i].max()], 
                [y_sample[:, i].min(), y_sample[:, i].max()], 'r--')
        plt.xlabel('实际占比')
        plt.ylabel('预测占比')
        plt.title(f'{mode_names[mode]} - 预测 vs 实际')
        
        plt.tight_layout()
        plt.savefig(f'shap_analysis_{mode}.png', dpi=300, bbox_inches='tight')
        plt.show()
        
        # 4. 详细的温度影响分析
        if temp_features:
            print(f"\n3. {mode_names[mode]} 温度影响的详细分析:")
            
            for idx, feature_name in temp_features:
                shap_vals = shap_values[:, idx]
                feature_vals = X_sample[:, idx]
                
                # 按特征值分组分析
                low_mask = feature_vals < np.percentile(feature_vals, 25)
                high_mask = feature_vals > np.percentile(feature_vals, 75)
                
                low_shap_mean = shap_vals[low_mask].mean()
                high_shap_mean = shap_vals[high_mask].mean()
                
                print(f"\n  {feature_name} 影响分析:")
                print(f"    低温情况下 (25%分位以下): 平均SHAP = {low_shap_mean:+.4f}")
                print(f"    高温情况下 (75%分位以上): 平均SHAP = {high_shap_mean:+.4f}")
                print(f"    温度效应差异: {high_shap_mean - low_shap_mean:+.4f}")
                
                # 判断温度效应类型
                if abs(high_shap_mean - low_shap_mean) > 0.01:
                    if high_shap_mean > low_shap_mean:
                        effect = f"高温促进{mode_names[mode]}出行"
                    else:
                        effect = f"高温抑制{mode_names[mode]}出行"
                else:
                    effect = f"温度对{mode_names[mode]}影响较小"
                
                print(f"    结论: {effect}")

def quick_temperature_analysis(model, X_test, y_test, feature_names):
    """快速分析温度特征影响"""
    
    if not install_shap_if_needed():
        print("无法进行SHAP分析，使用基础分析...")
        return
    
    import shap
    
    print("=== 快速温度影响分析 ===")
    
    modes = ['car', 'train', 'plane']
    n_samples = min(500, len(X_test))
    X_sample = X_test[:n_samples]
    
    # 找到温度特征
    temp_indices = [(i, name) for i, name in enumerate(feature_names) 
                   if 'temax' in name.lower()]
    
    for mode in modes:
        print(f"\n{mode} 占比 - 温度影响:")
        
        explainer = shap.TreeExplainer(model.models[mode])
        shap_values = explainer.shap_values(X_sample)
        
        for idx, temp_name in temp_indices:
            avg_impact = shap_values[:, idx].mean()
            direction = "正向" if avg_impact > 0 else "负向"
            magnitude = "强" if abs(avg_impact) > 0.01 else "中等" if abs(avg_impact) > 0.005 else "弱"
            
            print(f"  {temp_name}: {direction}影响, {magnitude} (SHAP均值: {avg_impact:+.4f})")


# ========== 第五部分：主程序 ==========
def main_training_and_analysis(transmod, df_imputed):
    """主要的训练和分析流程"""
    
    # 1. 训练模型（如果模型不存在）
    model_path = "transport_model_v1"
    
    # 创建模型实例
    share_model = TransportModeShareRegressor(
        model_type='xgboost',
        distance_file_path=r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\city_dist.csv'
    )
    
    # 尝试加载已保存的模型
    if not share_model.load_model(model_path):
        print("未找到保存的模型，开始训练新模型...")
        
        # 准备数据
        X_train, X_test, y_train, y_test = share_model.prepare_data(
            transmod,  # 你的训练数据
            use_enhanced_features=True
        )
        
        # 训练模型
        share_model.train(X_train, y_train)
        
        # 评估模型
        y_pred_test = share_model.evaluate(X_test, y_test)
        
        # 保存模型
        share_model.save_model(model_path)
        
        print("模型训练并保存完成!")
    else:
        print("成功加载已保存的模型!")
        
        # 重新准备测试数据用于分析
        X_train, X_test, y_train, y_test = share_model.prepare_data(
            transmod,
            use_enhanced_features=True
        )
    
    # 2. 进行SHAP分析
    print("\n开始SHAP分析...")
    
    feature_names = share_model.enhanced_features if share_model.enhanced_features else share_model.feature_names
    
    # 选择分析类型
    analysis_type = input("\n选择分析类型: 1-完整SHAP分析, 2-快速温度分析 (默认为2): ").strip()
    
    if analysis_type == "1":
        # 完整SHAP分析
        comprehensive_shap_analysis(
            share_model, X_test, y_test, feature_names, 
            sample_size=1000  # 可以调整样本数量
        )
    else:
        # 快速温度分析
        quick_temperature_analysis(share_model, X_test, y_test, feature_names)
    
    # 3. 进行预测（可选）
    do_prediction = input("\n是否进行2024年预测? (y/n, 默认n): ").strip().lower()
    
    if do_prediction == 'y':
        print("\n进行2024年预测...")
        predictions_2024 = predict_mode_shares(share_model, df_imputed)
        predictions_2024.to_csv('predictions_2024_with_shap_analysis.csv', index=False)
        
        # 分析结果
        analyze_beijing_shanghai_predictions(predictions_2024, transmod)
        print("预测结果已保存为: predictions_2024_with_shap_analysis.csv")
    
    return share_model, X_test, y_test, feature_names


# ========== 使用示例 ==========
if __name__ == "__main__":
    print("=== 交通模式占比预测与SHAP分析系统 ===")
    
    # 注意：这里需要你的数据变量 transmod 和 df_imputed
    # 如果没有这些数据，请先加载你的数据
    
    try:
        # 运行主程序 - 这里你需要传入你的真实数据
        model, X_test, y_test, feature_names = main_training_and_analysis(transmod, df_imputed)
        
        print("请确保你有以下数据变量:")
        print("- transmod: 训练数据")
        print("- df_imputed: 要预测的2024年数据")
        print("\n然后运行:")
        print("model, X_test, y_test, feature_names = main_training_and_analysis(transmod, df_imputed)")
        
        # 如果你想单独运行某个部分，可以使用以下代码：
        print("\n=== 单独功能示例 ===")
        print("1. 创建和训练模型:")
        print("   share_model = TransportModeShareRegressor()")
        print("   X_train, X_test, y_train, y_test = share_model.prepare_data(transmod)")
        print("   share_model.train(X_train, y_train)")
        
        print("\n2. 保存模型:")
        print("   share_model.save_model('my_model')")
        
        print("\n3. 加载模型:")
        print("   share_model.load_model('my_model')")
        
        print("\n4. SHAP分析:")
        print("   quick_temperature_analysis(share_model, X_test, y_test, feature_names)")
        
    except NameError as e:
        print(f"缺少数据变量: {e}")
        print("请先加载你的训练数据 (transmod) 和预测数据 (df_imputed)")

    print("\n=== 程序说明 ===")
    print("1. 首次运行会训练模型并自动保存")
    print("2. 后续运行会加载已保存的模型，节省时间")
    print("3. 支持完整SHAP分析和快速温度分析")
    print("4. 可选择是否进行预测")
    print("5. 生成的图片文件会保存在当前目录")
```

```python
import pandas as pd
import numpy as np

# 读取城市距离数据
city_dist = pd.read_csv(r'C:\Users\Fe6ruar9\PhD\313.5-Heatwave\city_dist.csv')

# 查看数据结构
print("城市距离数据结构：")
print(city_dist.head())
print(f"\n数据形状：{city_dist.shape}")

# 假设列名为 '城市1', '城市2', '距离(公里)'
# 如果实际列名不同，请相应修改
city1_col = '城市1'
city2_col = '城市2'
dist_col = '距离(公里)'

# 创建双向匹配的距离字典
# 由于城市对没有方向性，需要创建两个方向的映射
dist_dict = {}
for _, row in city_dist.iterrows():
    city1 = row[city1_col]
    city2 = row[city2_col]
    distance = row[dist_col]
    
    # 双向存储
    dist_dict[(city1, city2)] = distance
    dist_dict[(city2, city1)] = distance

# 将距离信息添加到new_df
def get_distance(row):
    origin = row['origin']
    dest = row['dest']
    return dist_dict.get((origin, dest), np.nan)

new_df['distance_km'] = new_df.apply(get_distance, axis=1)

# 检查是否有缺失的距离
missing_distances = new_df['distance_km'].isna().sum()
print(f"\n缺失距离的记录数：{missing_distances}")
if missing_distances > 0:
    print("缺失距离的城市对示例：")
    print(new_df[new_df['distance_km'].isna()][['origin', 'dest']].head())

# 计算实际出行人数
new_df['travel_population'] = new_df['travel_intensity'] * 32003.726712073505

# 计算各交通方式的出行人数
new_df['car_population'] = new_df['travel_population'] * new_df['car_share']
new_df['train_population'] = new_df['travel_population'] * new_df['train_share']
new_df['plane_population'] = new_df['travel_population'] * new_df['plane_share']

# 定义碳排放因子（单位：g CO2/人·公里）
carbon_factors = {
    'car': {'ave': 141, 'high': 208, 'low': 64},
    'train': {'ave': 19, 'high': 101, 'low': 6},
    'plane': {'ave': 123, 'high': 98, 'low': 133}
}

# 计算碳排放（单位：g CO2）
# 汽车碳排
new_df['car_emission_ave'] = new_df['car_population'] * new_df['distance_km'] * carbon_factors['car']['ave']
new_df['car_emission_high'] = new_df['car_population'] * new_df['distance_km'] * carbon_factors['car']['high']
new_df['car_emission_low'] = new_df['car_population'] * new_df['distance_km'] * carbon_factors['car']['low']

# 火车碳排
new_df['train_emission_ave'] = new_df['train_population'] * new_df['distance_km'] * carbon_factors['train']['ave']
new_df['train_emission_high'] = new_df['train_population'] * new_df['distance_km'] * carbon_factors['train']['high']
new_df['train_emission_low'] = new_df['train_population'] * new_df['distance_km'] * carbon_factors['train']['low']

# 飞机碳排
new_df['plane_emission_ave'] = new_df['plane_population'] * new_df['distance_km'] * carbon_factors['plane']['ave']
new_df['plane_emission_high'] = new_df['plane_population'] * new_df['distance_km'] * carbon_factors['plane']['high']
new_df['plane_emission_low'] = new_df['plane_population'] * new_df['distance_km'] * carbon_factors['plane']['low']

# 计算总碳排放（三种交通方式之和）
new_df['total_emission_ave'] = new_df['car_emission_ave'] + new_df['train_emission_ave'] + new_df['plane_emission_ave']
new_df['total_emission_high'] = new_df['car_emission_high'] + new_df['train_emission_high'] + new_df['plane_emission_high']
new_df['total_emission_low'] = new_df['car_emission_low'] + new_df['train_emission_low'] + new_df['plane_emission_low']

# 将碳排放从克转换为吨（可选）
emission_columns = [col for col in new_df.columns if 'emission' in col]
for col in emission_columns:
    new_df[f'{col}_tons'] = new_df[col] / 1e6  # 从克转换为吨

# 显示结果汇总
print("\n=== 碳排放计算结果汇总 ===")
print(f"\n总记录数：{len(new_df)}")
print(f"成功匹配距离的记录数：{len(new_df.dropna(subset=['distance_km']))}")

# 按交通方式汇总碳排放（单位：吨）
print("\n各交通方式总碳排放（吨）：")
print("平均碳排放因子情况：")
print(f"  汽车：{new_df['car_emission_ave_tons'].sum():.2f}")
print(f"  火车：{new_df['train_emission_ave_tons'].sum():.2f}")
print(f"  飞机：{new_df['plane_emission_ave_tons'].sum():.2f}")
print(f"  总计：{new_df['total_emission_ave_tons'].sum():.2f}")

print("\n高碳排放因子情况：")
print(f"  汽车：{new_df['car_emission_high_tons'].sum():.2f}")
print(f"  火车：{new_df['train_emission_high_tons'].sum():.2f}")
print(f"  飞机：{new_df['plane_emission_high_tons'].sum():.2f}")
print(f"  总计：{new_df['total_emission_high_tons'].sum():.2f}")

print("\n低碳排放因子情况：")
print(f"  汽车：{new_df['car_emission_low_tons'].sum():.2f}")
print(f"  火车：{new_df['train_emission_low_tons'].sum():.2f}")
print(f"  飞机：{new_df['plane_emission_low_tons'].sum():.2f}")
print(f"  总计：{new_df['total_emission_low_tons'].sum():.2f}")

# 可选：保存结果
# new_df.to_csv('carbon_emission_results.csv', index=False, encoding='utf-8-sig')

# 可选：查看前几行结果
print("\n结果示例（前5行）：")
display_cols = ['origin', 'dest', 'distance_km', 'travel_population', 
                'car_emission_ave_tons', 'train_emission_ave_tons', 'plane_emission_ave_tons', 
                'total_emission_ave_tons']
print(new_df[display_cols].head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.cm import ScalarMappable
from matplotlib.patches import Rectangle
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

# Set font to support English
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

def create_origin_grouped_scatter(merged_data, use_english_names=True):
    """
    Create a scatter plot of carbon emissions grouped by origin city
    with custom color gradient and highlighting for megacities.
    Shows total emissions instead of average emissions.
    Includes an inset plot for zoomed view of low-emission cities.
    """
    
    # Define megacity list
    megacity_list = ["上海", "北京", "深圳", "重庆", "广州", "成都", "天津", 
                     "武汉", "东莞", "西安", "杭州", "佛山", "南京", "沈阳", 
                     "青岛", "济南", "长沙", "哈尔滨", "郑州", "昆明", "大连"]
    
    # City name mapping (Chinese to English)
    city_name_map = {
        "上海": "Shanghai", "北京": "Beijing", "深圳": "Shenzhen", 
        "重庆": "Chongqing", "广州": "Guangzhou", "成都": "Chengdu",
        "天津": "Tianjin", "武汉": "Wuhan", "东莞": "Dongguan",
        "西安": "Xi'an", "杭州": "Hangzhou", "佛山": "Foshan",
        "南京": "Nanjing", "沈阳": "Shenyang", "青岛": "Qingdao",
        "济南": "Jinan", "长沙": "Changsha", "哈尔滨": "Harbin",
        "郑州": "Zhengzhou", "昆明": "Kunming", "大连": "Dalian",
        "苏州": "Suzhou", "宁波": "Ningbo", "无锡": "Wuxi",
        "厦门": "Xiamen", "福州": "Fuzhou", "合肥": "Hefei",
        "南昌": "Nanchang", "石家庄": "Shijiazhuang", "太原": "Taiyuan",
        "长春": "Changchun", "呼和浩特": "Hohhot", "南宁": "Nanning",
        "贵阳": "Guiyang", "海口": "Haikou", "兰州": "Lanzhou",
        "银川": "Yinchuan", "西宁": "Xining", "拉萨": "Lhasa",
        "乌鲁木齐": "Urumqi", "温州": "Wenzhou", "绍兴": "Shaoxing",
        "嘉兴": "Jiaxing", "常州": "Changzhou", "徐州": "Xuzhou",
        "南通": "Nantong", "扬州": "Yangzhou", "盐城": "Yancheng",
        "泰州": "Taizhou", "镇江": "Zhenjiang", "淮安": "Huai'an",
        "连云港": "Lianyungang", "宿迁": "Suqian", "台州": "Taizhou",
        "金华": "Jinhua", "湖州": "Huzhou", "衢州": "Quzhou",
        "舟山": "Zhoushan", "丽水": "Lishui",'伊犁哈萨克自治州':'Ili Kazakh','呼伦贝尔':'Hulunbuir'
    }
    
    # Group by origin and calculate total emissions (changed from mean to sum)
    grouped_data = merged_data.groupby('origin').agg({
        'routine_emission_ave_tons': 'sum',
        'heatwave_emission_ave_tons': 'sum',
        'diff_emission_ave_tons': 'sum'
    }).reset_index()
    
    # Rename columns to reflect that these are now totals
    grouped_data = grouped_data.rename(columns={
        'routine_emission_ave_tons': 'routine_emission_total_tons',
        'heatwave_emission_ave_tons': 'heatwave_emission_total_tons',
        'diff_emission_ave_tons': 'diff_emission_total_tons'
    })
    
    # Create display names column
    if use_english_names:
        grouped_data['display_name'] = grouped_data['origin'].map(city_name_map).fillna(grouped_data['origin'])
    else:
        grouped_data['display_name'] = grouped_data['origin']
    
    # Create figure with main plot
    fig = plt.figure(figsize=(15, 10))
    ax_main = fig.add_subplot(111)
    
    # Create custom colormap from #4F71BE to #C863A1
    colors = ['#4F71BE', '#C863A1']
    n_bins = 100
    cmap = mcolors.LinearSegmentedColormap.from_list('custom_gradient', colors, N=n_bins)
    
    # Normalize diff values for color mapping
    norm = plt.Normalize(vmin=grouped_data['diff_emission_total_tons'].min(), 
                        vmax=grouped_data['diff_emission_total_tons'].max())
    
    # Separate megacities and regular cities
    megacities_data = grouped_data[grouped_data['origin'].isin(megacity_list)]
    regular_cities_data = grouped_data[~grouped_data['origin'].isin(megacity_list)]
    
    # Plot on main axis
    scatter1 = ax_main.scatter(regular_cities_data['routine_emission_total_tons'], 
                              regular_cities_data['heatwave_emission_total_tons'],
                              c=regular_cities_data['diff_emission_total_tons'],
                              cmap=cmap, norm=norm,
                              s=200, alpha=0.7, edgecolors='white', linewidth=0.5,
                              marker='o', label='Other Cities')
    
    scatter2 = ax_main.scatter(megacities_data['routine_emission_total_tons'], 
                            megacities_data['heatwave_emission_total_tons'],
                            color='#FFA500',
                             s=1000, alpha=0.9, edgecolors='white', linewidth=0.5,
                             marker='*', label='Megacities')
    
    # Add diagonal reference line
    max_val = max(grouped_data['routine_emission_total_tons'].max(), 
                  grouped_data['heatwave_emission_total_tons'].max())
    min_val = min(grouped_data['routine_emission_total_tons'].min(), 
                  grouped_data['heatwave_emission_total_tons'].min())
    ax_main.plot([min_val, max_val], [min_val, max_val], 
                'k--', alpha=0.3, linewidth=1, label='No Change Line')
    
    # Add colorbar
    sm = ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = plt.colorbar(sm, ax=ax_main, pad=0.02)
    cbar.set_label('Total Emission Difference (tons)\n(Other Cities Only)', fontsize=12)
    
    # Set labels and title for main plot (updated to reflect totals)
    ax_main.set_xlabel('Total Routine Emissions (tons)', fontsize=14)
    ax_main.set_ylabel('Total Heatwave Emissions (tons)', fontsize=14)
    ax_main.set_title('Carbon Emissions: Total Heatwave vs Routine Period by Origin City', 
                      fontsize=16, fontweight='bold', pad=20)
    
    # Add grid
    ax_main.grid(True, alpha=0.3, linestyle='--')
    
    # Add legend
    ax_main.legend(loc='upper left', frameon=True, fancybox=True, shadow=True)
    
    # Adjust zoom area based on new scale (since totals will be larger)
    # You may need to adjust these values based on your data
# 将zoom_threshold改为固定值2000
    zoom_threshold = 2000  # 固定缩放范围为0-2000
    
    # Add rectangle to show zoomed area
    zoom_rect = Rectangle((0, 0), zoom_threshold, zoom_threshold, linewidth=2, edgecolor='red', 
                         facecolor='none', linestyle='--', alpha=0.8)
    ax_main.add_patch(zoom_rect)
    
    # Create inset axes for zoomed view
    ax_inset = inset_axes(ax_main, width="35%", height="35%", loc='lower right',
                         bbox_to_anchor=(0, 0.05, 1, 1), bbox_transform=ax_main.transAxes)
    
    # Filter data for zoom view (changed to fixed threshold of 2000)
    zoom_data = grouped_data[(grouped_data['routine_emission_total_tons'] <= zoom_threshold) & 
                            (grouped_data['heatwave_emission_total_tons'] <= zoom_threshold)]
    zoom_megacities = zoom_data[zoom_data['origin'].isin(megacity_list)]
    zoom_regular = zoom_data[~zoom_data['origin'].isin(megacity_list)]
    
    # Plot on inset axis
    ax_inset.scatter(zoom_regular['routine_emission_total_tons'], 
                    zoom_regular['heatwave_emission_total_tons'],
                    c=zoom_regular['diff_emission_total_tons'],
                    cmap=cmap, norm=norm,
                    s=60, alpha=0.7, edgecolors='white', linewidth=0.5,
                    marker='o')
    
    ax_inset.scatter(zoom_megacities['routine_emission_total_tons'], 
                    zoom_megacities['heatwave_emission_total_tons'],
                    color='#FFA500',
                    s=300, alpha=0.9, edgecolors='white', linewidth=1,
                    marker='*')
    
    # Add diagonal line to inset
    ax_inset.plot([0, zoom_threshold], [0, zoom_threshold], 'k--', alpha=0.3, linewidth=1)
    
    # Set inset limits and labels with fixed range
    ax_inset.set_xlim(0, zoom_threshold)
    ax_inset.set_ylim(0, zoom_threshold)
    ax_inset.set_xlabel('Routine (tons)', fontsize=10)
    ax_inset.set_ylabel('Heatwave (tons)', fontsize=10)
    ax_inset.set_title('Zoomed View (0-2000 tons)', fontsize=11, fontweight='bold')
    ax_inset.grid(True, alpha=0.3, linestyle='--')
    
    # Add labels for some cities in the zoomed view
    zoom_top = zoom_data.nlargest(5, 'diff_emission_total_tons')
    for i, (_, city) in enumerate(zoom_top.iterrows()):
        # Smart offset for zoomed view
        x_pos = city['routine_emission_total_tons']
        y_pos = city['heatwave_emission_total_tons']
        
        # Distribute annotations around the plot
        angles = [45, -45, 135, -135, 90]
        angle = angles[i % len(angles)]
        radius = zoom_threshold * 0.1  # Adjust radius based on zoom threshold
        x_offset = radius * np.cos(np.radians(angle))
        y_offset = radius * np.sin(np.radians(angle))
            
        ax_inset.annotate(city['display_name'], 
                         xy=(x_pos, y_pos),
                         xytext=(x_offset, y_offset), textcoords='offset points',
                         fontsize=8, fontweight='normal',
                         arrowprops=dict(arrowstyle='->', 
                                       connectionstyle='arc3,rad=0.15',
                                       color='black',
                                       lw=0.8,
                                       alpha=0.7))
        
        # Limit to 3 annotations
        if i >= 2:
            break
    
    # Add annotation for top emission increases on main plot
    top_cities = grouped_data.nlargest(5, 'diff_emission_total_tons')
    annotated_cities = []
    for i, (_, city) in enumerate(top_cities.iterrows()):
        # Only annotate if not in the zoom area
        if city['routine_emission_total_tons'] > zoom_threshold or city['heatwave_emission_total_tons'] > zoom_threshold:
            # Smart offset calculation to avoid overlap
            x_pos = city['routine_emission_total_tons']
            y_pos = city['heatwave_emission_total_tons']
            
            # Determine quadrant and set offset accordingly
            if x_pos > grouped_data['routine_emission_total_tons'].median():
                x_offset = 40 if i % 2 == 0 else 50
            else:
                x_offset = -40 if i % 2 == 0 else -50
                
            if y_pos > grouped_data['heatwave_emission_total_tons'].median():
                y_offset = 30 if i % 2 == 0 else -20
            else:
                y_offset = -30 if i % 2 == 0 else 20
            
            ax_main.annotate(city['display_name'], 
                           xy=(x_pos, y_pos),
                           xytext=(x_offset, y_offset), textcoords='offset points',
                           fontsize=10, fontweight='bold',
                           arrowprops=dict(arrowstyle='->', 
                                         connectionstyle='arc3,rad=0.2',
                                         color='black',
                                         lw=1.2,
                                         alpha=0.8))
            annotated_cities.append(city['display_name'])
            
            # Limit to 3 annotations to avoid clutter
            if len(annotated_cities) >= 3:
                break
    
    # Adjust layout
    plt.tight_layout()
    
    # Print summary statistics (updated for totals)
    print("Summary Statistics by Origin City (Total Emissions):")
    print("-" * 50)
    print(f"Total cities analyzed: {len(grouped_data)}")
    print(f"Megacities in dataset: {len(megacities_data)}")
    print(f"Cities in zoomed view: {len(zoom_data)}")
    print(f"Total emission increase across all cities: {grouped_data['diff_emission_total_tons'].sum():.2f} tons")
    print(f"Average emission increase per city: {grouped_data['diff_emission_total_tons'].mean():.2f} tons")
    print(f"Cities with increased emissions: {(grouped_data['diff_emission_total_tons'] > 0).sum()}")
    print("-" * 50)

# Print top 5 cities with highest emission increases
    print("\nTop 5 Cities with Highest Total Emission Increases:")
    top_5 = grouped_data.nlargest(5, 'diff_emission_total_tons')[['origin', 'display_name', 'diff_emission_total_tons']]
    for idx, (_, row) in enumerate(top_5.iterrows(), 1):
        mega_indicator = " (Megacity)" if row['origin'] in megacity_list else ""
        print(f"{idx}. {row['display_name']}{mega_indicator}: +{row['diff_emission_total_tons']:.2f} tons")
    
    return fig, grouped_data

# Usage example
if __name__ == "__main__":
    # Create the scatter plot with English city names
    fig, grouped_data = create_origin_grouped_scatter(merged_data, use_english_names=True)
    
    # Save the figure
    plt.savefig('origin_city_total_emissions_scatter.pdf', dpi=300, bbox_inches='tight')
    plt.show()
    
    # Optional: Export grouped data for further analysis
    # grouped_data.to_csv('grouped_total_emissions_by_origin.csv', index=False)
    
    # Alternative: Use Chinese names (if your system supports Chinese fonts)
    # fig2, grouped_data2 = create_origin_grouped_scatter(merged_data, use_english_names=False)
```

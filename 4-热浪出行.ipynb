```python
data_intensity_2024 = pd.read_csv('data_intensity_2024.csv', index_col=None)
heatwave_periods_2024 = pd.read_csv('heatwave_periods_2024.csv', index_col=None)
```

```python
data_intensity_2024['城市'] = data_intensity_2024['城市'].map(replacement_dict).fillna(data_intensity_2024['城市'])
data_intensity_2024['目的城市'] = data_intensity_2024['目的城市'].map(replacement_dict).fillna(data_intensity_2024['目的城市'])
heatwave_periods_2024['城市'] = heatwave_periods_2024['城市'].map(replacement_dict).fillna(heatwave_periods_2024['城市'])
```

```python
data_intensity_2024['日期'] = pd.to_datetime(data_intensity_2024['日期'])
heatwave_periods_2024['开始日期'] = pd.to_datetime(heatwave_periods_2024['开始日期'])
heatwave_periods_2024['结束日期'] = pd.to_datetime(heatwave_periods_2024['结束日期'])
data_intensity_2024['day_index'] = (data_intensity_2024['日期'] - pd.to_datetime('2024-01-01')).dt.days
```

```python
# 定义函数，根据城市名称和天数索引返回热浪编码
def get_heatwave(city, day_index):
    # 如果字典中有该城市
    if city in city_heatwave_encoding_2024:
        return city_heatwave_encoding_2024[city][day_index]
    else:
        return None
```

```python
# 使用 apply 方法为出发城市（'城市'）和目的城市（'目的城市'）获取对应的热浪编码
data_intensity_2024['is_heatwave_o'] = data_intensity_2024.apply(lambda row: get_heatwave(row['城市'], row['day_index']), axis=1)
data_intensity_2024['is_heatwave_d'] = data_intensity_2024.apply(lambda row: get_heatwave(row['目的城市'], row['day_index']), axis=1)
```

```python
data_intensity_2024
```

```python
data_intensity_2024["HE"] = ((data_intensity_2024["is_heatwave_o"] == 1) & (data_intensity_2024["is_heatwave_d"] == 0)).astype(int)
data_intensity_2024["HT"] = ((data_intensity_2024["is_heatwave_o"] == 1) & (data_intensity_2024["is_heatwave_d"] == 1)).astype(int)
data_intensity_2024["HEC"] = ((data_intensity_2024["is_heatwave_o"] == 0) & (data_intensity_2024["is_heatwave_d"] == 1)).astype(int)
data_intensity_2024["Routine"] = ((data_intensity_2024["is_heatwave_o"] == 0) & (data_intensity_2024["is_heatwave_d"] == 0)).astype(int)
```
